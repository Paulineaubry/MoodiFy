```python
import pandas as pd

def creer_colonnes_tags_humeur_activite(df_final_corrigé):
    """
    Fonction qui crée deux colonnes 'Tag_Humeur' et 'Tag_Activité' dans le DataFrame
    en appliquant toutes les fonctions de tagging, supprimant les doublons,
    et nettoyant les colonnes intermédiaires.
    
    Arguments :
        df : pandas.DataFrame, le DataFrame contenant les colonnes audio features
        
    Retourne :
        pandas.DataFrame : le DataFrame avec les deux nouvelles colonnes et 
                          les colonnes intermédiaires supprimées
    """
    
    # ========== FONCTIONS DE TAGGING CORRIGÉES ==========
    
    def valence_tag_humeur(valence):
        """Attribue un tag d'humeur basé sur la valence (positivité musicale)"""
        if valence >= 0.5:
            return "#joyeux"
        else:
            return "#triste"
    
    def mode_tag_humeur(mode):
        """Attribue un tag d'humeur basé sur le mode (majeur/mineur)"""
        if mode == 1:
            return "#joyeux"
        else:
            return "#triste"
    
    def acousticness_tag_humeur(acousticness):
        """Attribue un tag d'humeur basé sur l'acousticité"""
        if acousticness >= 0.5:
            return "#calme"
        else:
            return "#energique"
    
    def loudness_tag_humeur(loudness):
        """Attribue un tag d'humeur basé sur le volume sonore"""
        if loudness >= -30:
            return "#energique"
        else:
            return "#calme"
    
    def danceability_tag_activite(danceability):
        """Attribue un tag d'activité basé sur la dansabilité"""
        if danceability >= 0.5:
            return "#dansant #fête"
        else:
            return "#concentration #travail"
    
    def energy_tag_activite(energy):
        """Attribue un tag d'activité basé sur l'énergie"""
        if energy >= 0.8:
            return "#cardio"
        elif 0.8 > energy >= 0.4:
            return "#sport"
        else:
            return "#meditation #yoga"
    
    def instrumentalness_tag_activite(instrumentalness):
        """Attribue un tag d'activité basé sur l'instrumentalité"""
        if instrumentalness >= 0.5:
            return "#concentration #meditation #relaxation"
        else:
            return "#sport #cardio"
    
    def tempo_tag_activite(tempo):
        """Attribue un tag d'activité basé sur le tempo"""
        if tempo >= 180:
            return "#cardio"
        elif 180 > tempo >= 120:
            return "#sport"
        elif 120 > tempo >= 80:
            return "#energique"
        elif 80 > tempo > 60:
            return "#calme"
        else:
            return "#meditation #yoga #sommeil"
    
    # ========== APPLICATION DES FONCTIONS ==========
    
    # Application des fonctions d'humeur - création des colonnes intermédiaires
    df_final_corrigé['valence_humeur'] = df_final_corrigé['valence'].apply(valence_tag_humeur)
    df_final_corrigé['mode_humeur'] = df_final_corrigé['mode'].apply(mode_tag_humeur)
    df_final_corrigé['acousticness_humeur'] = df_final_corrigé['acousticness'].apply(acousticness_tag_humeur)
    df_final_corrigé['loudness_humeur'] = df_final_corrigé['loudness'].apply(loudness_tag_humeur)
    
    # Application des fonctions d'activité - création des colonnes intermédiaires
    df_final_corrigé['danceability_activite'] = df_final_corrigé['danceability'].apply(danceability_tag_activite)
    df_final_corrigé['energy_activite'] = df_final_corrigé['energy'].apply(energy_tag_activite)
    df_final_corrigé['instrumentalness_activite'] = df_final_corrigé['instrumentalness'].apply(instrumentalness_tag_activite)
    df_final_corrigé['tempo_activite'] = df_final_corrigé['tempo'].apply(tempo_tag_activite)
    
    # ========== DÉFINITION DES COLONNES À FUSIONNER ==========
    
    # Colonnes contenant les tags d'humeur (colonnes intermédiaires)
    colonnes_humeur = [
        'valence_humeur',
        'mode_humeur',
        'acousticness_humeur',
        'loudness_humeur'
    ]
    
    # Colonnes contenant les tags d'activité (colonnes intermédiaires)
    colonnes_activite = [
        'danceability_activite',
        'energy_activite',
        'instrumentalness_activite',
        'tempo_activite'
    ]
    
    # ========== FONCTION DE FUSION AVEC SUPPRESSION DES DOUBLONS ==========
    
    def fusionner_tags_sans_doublons(row):
        """
        Fusionne les tags d'une ligne en supprimant les doublons.
        
        Arguments :
            row : pandas.Series, une ligne du DataFrame
            
        Retourne :
            str : chaîne de tags uniques, triés, séparés par un espace
        """
        # Liste pour stocker tous les tags individuels
        tous_les_tags = []
        
        # Parcourt chaque valeur de la ligne (chaque colonne de tag)
        for valeur in row:
            # Vérifie si la valeur n'est pas nulle
            if pd.notnull(valeur):
                # Divise la chaîne de tags en tags individuels (séparés par des espaces)
                tags_individuels = str(valeur).split()
                # Ajoute chaque tag à la liste globale
                tous_les_tags.extend(tags_individuels)
        
        # Convertit la liste en set pour supprimer les doublons, puis remet en liste
        tags_uniques = list(set(tous_les_tags))
        
        # Trie les tags par ordre alphabétique pour une présentation cohérente
        tags_uniques.sort()
        
        # Joint tous les tags uniques avec un espace comme séparateur
        return ' '.join(tags_uniques)
    
    # ========== CRÉATION DES COLONNES FINALES ==========
    
    # Création de la colonne Tag_Humeur en fusionnant les colonnes d'humeur
    df_final_corrigé['Tag_Humeur'] = df_final_corrigé[colonnes_humeur].apply(fusionner_tags_sans_doublons, axis=1)
    
    # Création de la colonne Tag_Activité en fusionnant les colonnes d'activité
    df_final_corrigé['Tag_Activité'] = df_final_corrigé[colonnes_activite].apply(fusionner_tags_sans_doublons, axis=1)
    
    # ========== SUPPRESSION DES COLONNES INTERMÉDIAIRES ==========
    
    # Combinaison de toutes les colonnes intermédiaires à supprimer
    colonnes_a_supprimer = colonnes_humeur + colonnes_activite
    
    # Suppression des colonnes intermédiaires du DataFrame
    # drop() supprime les colonnes spécifiées
    # columns= spécifie les noms des colonnes à supprimer
    # inplace=True modifie directement le DataFrame sans créer de copie
    df_final_corrigé.drop(columns=colonnes_a_supprimer, inplace=True)
    
    # Message de confirmation du nettoyage
    print(f"Colonnes intermédiaires supprimées : {colonnes_a_supprimer}")
    
    # Retourne le DataFrame modifié et nettoyé
    return df_final_corrigé

# ========== UTILISATION DE LA FONCTION ==========

# Application de la fonction sur ton DataFrame
df_final_corrigé = creer_colonnes_tags_humeur_activite(df_final_corrigé)

# Affichage des résultats pour vérification
print("Aperçu des nouvelles colonnes créées :")
print(df_final_corrigé[['Tag_Humeur', 'Tag_Activité']].head())

# Vérification que les colonnes intermédiaires ont bien été supprimées
print(f"\nNombre total de colonnes dans le DataFrame : {len(df_final_corrigé.columns)}")
print(f"Colonnes restantes : {list(df_final_corrigé.columns)}")

# Vérification des tags uniques générés
print("\nTags d'humeur uniques dans le dataset :")
tous_tags_humeur = set()
for tags_ligne in df_final_corrigé['Tag_Humeur']:
    tous_tags_humeur.update(tags_ligne.split())
print(sorted(tous_tags_humeur))

print("\nTags d'activité uniques dans le dataset :")
tous_tags_activite = set()
for tags_ligne in df_final_corrigé['Tag_Activité']:
    tous_tags_activite.update(tags_ligne.split())
print(sorted(tous_tags_activite))

```

```python
def creer_tags_manuel(df_final_corrigé: pd.DataFrame) -> pd.DataFrame:
    """
    Ajoute deux colonnes :
      - tags_humeur  : #joyeux, #triste, #energique, #calme
      - tags_activité: #dance/#fête, #concentration/#travail,
                       #sport/#cardio, #meditation/#yoga
    """
    # Définition du tag d'humeur
    def tag_humeur(valence, acousticness, loudness):
        if valence >= 0.75:
            return "#joyeux"
        if valence < 0.25:
            return "#triste"
        if loudness >= -20 or acousticness < 0.3:
            return "#energique"
        return "#calme"

    # Définition du tag d'activité
    def tag_activite(danceability, energy, tempo):
        if danceability >= 0.5:
            return "#dance/#fête"
        if energy >= 0.6 or tempo >= 110:
            return "#sport/#cardio"
        if energy >= 0.3:
            return "#concentration/#travail"
        return "#meditation/#yoga"

    # Application ligne par ligne
    df_final_corrigé["tags_humeur"] = df_final_corrigé.apply(
        lambda r: tag_humeur(r["valence"], r["acousticness"], r["loudness"]), axis=1
    )
    df_final_corrigé["tags_activité"] = df_final_corrigé.apply(
        lambda r: tag_activite(r["danceability"], r["energy"], r["tempo"]), axis=1
    )

    # Vérification des tags uniques et comptages
    uniques_humeur = df_final_corrigé["tags_humeur"].unique().tolist()
    compte_humeur = df_final_corrigé["tags_humeur"].value_counts().reset_index()
    compte_humeur.columns = ["tag_humeur", "occurrences"]

    uniques_activite = df_final_corrigé["tags_activité"].unique().tolist()
    compte_activite = df_final_corrigé["tags_activité"].value_counts().reset_index()
    compte_activite.columns = ["tag_activité", "occurrences"]

    # Affichage des vérifications
    print("Tags d'humeur uniques :", uniques_humeur)
    print("\nRépartition des tags d'humeur :")
    print(compte_humeur)

    print("\nTags d'activité uniques :", uniques_activite)
    print("\nRépartition des tags d'activité :")
    print(compte_activite)

    return df_final_corrigé


df_final_corrigé_tagged = creer_tags_manuel(df_final_corrigé)
df_final_corrigé_tagged
```
