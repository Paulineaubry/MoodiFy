```python

import pandas as pd
import numpy as np

def identifier_artistes_variete_francaise(df_spotify):
    """
    Identifie les artistes de variété française dans un DataFrame Spotify
    
    Arguments:
    df_spotify (DataFrame): DataFrame contenant les données Spotify avec colonnes 
                           'artist_name', 'genre', 'track_name', etc.
    
    Retourne:
    DataFrame: DataFrame filtré contenant uniquement les artistes de variété française
    """
    
    # Liste des artistes de variété française reconnus
    # Basée sur les artistes mentionnés dans les sources [9][15][17]
    artistes_variete_francaise = [
        'Mylène Farmer', 'Johnny Hallyday', 'Claude François', 'Céline Dion',
        'Edith Piaf', 'Charles Aznavour', 'Daniel Balavoine', 'France Gall',
        'Jacques Brel', 'Stromae', 'Christophe Maé', 'Julien Doré',
        'Indochine', 'M. Pokora', 'Vanessa Paradis', 'Florent Pagny',
        'Renaud', 'Alain Souchon', 'Michel Sardou', 'Jean-Jacques Goldman',
        'Louane', 'Angèle', 'Christine and the Queens', 'Zaz', 'Zazie',
        'Grand corps malade', 'Jenifer', 'Garou', 'Patrick Bruel'
    ]
    
    # Conversion en minuscules pour faciliter les comparaisons
    # La fonction lower() normalise les chaînes pour éviter les problèmes de casse
    artistes_normalises = [artiste.lower() for artiste in artistes_variete_francaise]
    
    # Création d'une copie du DataFrame pour éviter les modifications inattendues
    # La méthode copy() crée une copie indépendante du DataFrame original
    df_filtre = df_spotify.copy()
    
    # Normalisation des noms d'artistes dans le DataFrame
    # La méthode str.lower() applique la conversion en minuscules à toute la série
    df_filtre['artist_name_lower'] = df_filtre['artist_name'].str.lower()
    
    # Filtrage principal par noms d'artistes exacts
    # La fonction isin() retourne un masque booléen pour les valeurs présentes dans la liste
    masque_artistes = df_filtre['artist_name_lower'].isin(artistes_normalises)
    
    # Filtrage secondaire par genre si la colonne existe
    # Vérification de l'existence de la colonne 'genre' pour éviter les erreurs
    if 'genre' in df_filtre.columns:
        # La méthode str.contains() avec case=False pour recherche insensible à la casse
        # Le paramètre na=False traite les valeurs manquantes comme False
        masque_genre_francais = df_filtre['genre'].str.contains(
            'french|français|francophone|variété française|chanson française', 
            case=False, 
            na=False,
            regex=True  # Permet l'utilisation d'expressions régulières
        )
        
        # Combinaison des masques avec l'opérateur OR (|)
        # Permet de capturer les artistes identifiés par nom OU par genre
        masque_final = masque_artistes | masque_genre_francais
    else:
        # Si pas de colonne genre, utilisation uniquement du masque artistes
        masque_final = masque_artistes
    
    # Application du masque pour filtrer le DataFrame
    # L'indexation booléenne retourne uniquement les lignes où le masque est True
    resultats = df_filtre[masque_final]
    
    # Filtrage complémentaire par mots-clés dans les titres de chansons
    # Recherche de termes français typiques dans les titres
    mots_cles_francais = ['amour', 'coeur', 'vie', 'monde', 'temps', 'jour', 'nuit']
    
    # Création d'un pattern regex pour rechercher les mots-clés français
    # La fonction join() crée une expression régulière avec OR (|) entre les mots
    pattern_francais = '|'.join(mots_cles_francais)
    
    # Recherche dans les titres de chansons si la colonne existe
    if 'track_name' in df_filtre.columns:
        # str.contains() avec les mêmes paramètres que précédemment
        masque_titres_francais = df_filtre['track_name'].str.contains(
            pattern_francais,
            case=False,
            na=False,
            regex=True
        )
        
        # Ajout des résultats du filtrage par titres
        # Combinaison avec les résultats précédents via concat()
        resultats_titres = df_filtre[masque_titres_francais & ~masque_final]
        
        # concat() combine les DataFrames verticalement
        # ignore_index=True recréé un index continu
        resultats = pd.concat([resultats, resultats_titres], ignore_index=True)
    
    # Suppression des doublons potentiels
    # drop_duplicates() élimine les lignes identiques
    # subset=['artist_name', 'track_name'] spécifie les colonnes à considérer
    resultats = resultats.drop_duplicates(subset=['artist_name', 'track_name'])
    
    # Tri par nom d'artiste pour faciliter la lecture
    # sort_values() organise le DataFrame selon la colonne spécifiée
    resultats = resultats.sort_values('artist_name')
    
    # Suppression de la colonne temporaire créée pour le filtrage
    # drop() retire les colonnes spécifiées du DataFrame
    if 'artist_name_lower' in resultats.columns:
        resultats = resultats.drop('artist_name_lower', axis=1)
    
    # Remise à zéro de l'index pour un affichage propre
    # reset_index() recrée un index numérique continu
    resultats = resultats.reset_index(drop=True)
    
    return resultats


def analyser_caracteristiques_audio_francaises(df_spotify_francais):
    """
    Analyse les caractéristiques audio spécifiques aux artistes français
    
    Arguments:
    df_spotify_francais (DataFrame): DataFrame des artistes français identifiés
    
    Retourne:
    dict: Dictionnaire des statistiques descriptives
    """
    
    # Colonnes des caractéristiques audio selon la documentation Spotify [5][7]
    caracteristiques_audio = [
        'danceability', 'energy', 'speechiness', 'acousticness',
        'instrumentalness', 'liveness', 'valence', 'tempo', 'duration_ms'
    ]
    
    # Sélection des colonnes audio disponibles
    # La compréhension de liste filtre les colonnes existantes dans le DataFrame
    colonnes_disponibles = [col for col in caracteristiques_audio 
                           if col in df_spotify_francais.columns]
    
    # Calcul des statistiques descriptives
    # describe() génère les statistiques (moyenne, médiane, écart-type, etc.)
    if colonnes_disponibles:
        statistiques = df_spotify_francais[colonnes_disponibles].describe()
        
        # Conversion en dictionnaire pour un accès plus facile
        # to_dict() transforme le DataFrame de statistiques en dictionnaire
        stats_dict = statistiques.to_dict()
        
        return stats_dict
    else:
        # Retour d'un dictionnaire vide si aucune caractéristique n'est disponible
        return {}


def filtrer_par_popularite_francaise(df_spotify_francais, seuil_popularite=50):
    """
    Filtre les artistes français selon leur popularité
    
    Arguments:
    df_spotify_francais (DataFrame): DataFrame des artistes français
    seuil_popularite (int): Seuil minimal de popularité (0-100)
    
    Retourne:
    DataFrame: Artistes français populaires
    """
    
    # Vérification de l'existence de la colonne popularité
    if 'popularity' not in df_spotify_francais.columns:
        print("Colonne 'popularity' non trouvée dans le DataFrame")
        return df_spotify_francais
    
    # Filtrage par seuil de popularité
    # L'opérateur >= crée un masque booléen pour les valeurs supérieures au seuil
    masque_popularite = df_spotify_francais['popularity'] >= seuil_popularite
    
    # Application du filtre
    artistes_populaires = df_spotify_francais[masque_popularite]
    
    # Tri par popularité décroissante
    # ascending=False classe du plus populaire au moins populaire
    artistes_populaires = artistes_populaires.sort_values(
        'popularity', 
        ascending=False
    )
    
    return artistes_populaires


def generer_rapport_variete_francaise(df_spotify):
    """
    Génère un rapport complet sur les artistes de variété française
    
    Arguments:
    df_spotify (DataFrame): DataFrame Spotify complet
    
    Retourne:
    dict: Rapport détaillé avec statistiques et listes
    """
    
    # Application de la fonction principale d'identification
    artistes_francais = identifier_artistes_variete_francaise(df_spotify)
    
    # Calcul des métriques de base
    # len() retourne le nombre de lignes du DataFrame
    nombre_total_artistes = len(artistes_francais['artist_name'].unique())
    nombre_total_titres = len(artistes_francais)
    
    # Analyse des caractéristiques audio
    stats_audio = analyser_caracteristiques_audio_francaises(artistes_francais)
    
    # Identification des artistes les plus populaires
    top_artistes_populaires = filtrer_par_popularite_francaise(
        artistes_francais, 
        seuil_popularite=70
    )
    
    # Création du rapport sous forme de dictionnaire
    rapport = {
        'nombre_artistes_uniques': nombre_total_artistes,
        'nombre_total_titres': nombre_total_titres,
        'artistes_les_plus_populaires': top_artistes_populaires.head(10),
        'statistiques_audio': stats_audio,
        'dataset_complet': artistes_francais
    }
    
    return rapport


# Exemple d'utilisation avec chargement de données
def exemple_utilisation():
    """
    Exemple d'utilisation des fonctions avec un DataFrame Spotify
    """
    
    # Chargement du DataFrame (remplacer par le vrai chemin du fichier)
    # read_csv() charge un fichier CSV en DataFrame pandas
    try:
        df_spotify = pd.read_csv('spotify_dataset.csv')
        
        # Application de l'analyse complète
        rapport_final = generer_rapport_variete_francaise(df_spotify)
        
        # Affichage des résultats principaux
        print(f"Nombre d'artistes français identifiés: {rapport_final['nombre_artistes_uniques']}")
        print(f"Nombre total de titres: {rapport_final['nombre_total_titres']}")
        
        # Affichage des 5 premiers artistes populaires
        if not rapport_final['artistes_les_plus_populaires'].empty:
            print("\nTop 5 des artistes les plus populaires:")
            print(rapport_final['artistes_les_plus_populaires'][['artist_name', 'popularity']].head())
        
        return rapport_final
        
    except FileNotFoundError:
        print("Fichier de données Spotify non trouvé")
        return None
    except Exception as e:
        print(f"Erreur lors du traitement: {e}")
        return None
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ... (conserve les fonctions précédentes telles quelles jusqu'à generer_rapport_variete_francaise)

def generer_rapport_variete_francaise(df_spotify, afficher_graphs=True):
    """
    Génère un rapport complet sur les artistes de variété française avec visualisation.
    
    Arguments:
    df_spotify (DataFrame): DataFrame Spotify complet
    afficher_graphs (bool): Affiche les visualisations si True
    
    Retourne:
    dict: Rapport détaillé avec statistiques et listes
    """
    
    # Application de la fonction principale d'identification
    artistes_francais = identifier_artistes_variete_francaise(df_spotify)
    
    # Calcul des métriques de base
    nombre_total_artistes = len(artistes_francais['artist_name'].unique())
    nombre_total_titres = len(artistes_francais)
    
    # Analyse des caractéristiques audio
    stats_audio = analyser_caracteristiques_audio_francaises(artistes_francais)
    
    # Identification des artistes les plus populaires
    top_artistes_populaires = filtrer_par_popularite_francaise(
        artistes_francais, 
        seuil_popularite=70
    ).head(10)

    if afficher_graphs and not top_artistes_populaires.empty:
        # 🌟 1. Bar chart : popularité des artistes
        plt.figure(figsize=(10, 5))
        sns.barplot(
            x='popularity', 
            y='artist_name', 
            data=top_artistes_populaires,
            palette='mako'
        )
        plt.title("🎤 Top 10 artistes français les plus populaires")
        plt.xlabel("Popularité Spotify (0-100)")
        plt.ylabel("Artiste")
        plt.tight_layout()
        plt.show()

    if afficher_graphs and stats_audio:
        # 🌟 2. Heatmap des caractéristiques audio moyennes
        moyennes = {k: v['mean'] for k, v in stats_audio.items()}
        df_moyennes = pd.DataFrame(moyennes, index=['Valeur moyenne'])
        
        plt.figure(figsize=(10, 2))
        sns.heatmap(df_moyennes, annot=True, cmap='YlGnBu', fmt=".2f", cbar=True)
        plt.title("🎼 Moyennes des caractéristiques audio (artistes français)")
        plt.tight_layout()
        plt.show()
    
    # Création du rapport final
    rapport = {
        'nombre_artistes_uniques': nombre_total_artistes,
        'nombre_total_titres': nombre_total_titres,
        'artistes_les_plus_populaires': top_artistes_populaires,
        'statistiques_audio': stats_audio,
        'dataset_complet': artistes_francais
    }
    
    return rapport
```

```python
def detecter_variete_francaise(df):
    # Copie de sécurité
    df = df.copy()

    # On remplace les NaN par des chaînes vides (important !)
    df['artist_name'] = df['artist_name'].fillna('')
    df['track_name'] = df['track_name'].fillna('')
    df['genre'] = df['genre'].fillna('')

    # Normalisation des textes
    df['artist_name_lower'] = df['artist_name'].str.lower()
    df['track_name_lower'] = df['track_name'].str.lower()
    df['genre_lower'] = df['genre'].str.lower()

    # A. Genres typiques
    genres_cibles = ['chanson française', 'french pop', 'variété française', 'francophone']
    genre_match = df['genre_lower'].apply(lambda g: any(x in g for x in genres_cibles))

    # B. Mots-clés dans les titres
    mots_cles = ['amour', 'coeur', 'liberté', 'vie', 'rêve', 'toi', 'moi', 'nous', 'ensemble', 'souvenir']
    mots_match = df['track_name_lower'].apply(lambda t: any(mot in t for mot in mots_cles))

    # C. Artistes connus
    artistes_variete_fr = [
        "charles aznavour", "julien doré", "zaz", "vianney", "jean-jacques goldman",
        "edith piaf", "francis cabrel", "clara luciani", "coeur de pirate", "florent pagny"
    ]
    artist_match = df['artist_name_lower'].isin(artistes_variete_fr)

    # Fusion
    filtre_global = genre_match | mots_match | artist_match

    return df[filtre_global].copy()
```

```python
def scorer_variete_francaise(df):
    df = df.copy()
    df['artist_name'] = df['artist_name'].fillna('')
    df['track_name'] = df['track_name'].fillna('')
    df['genre'] = df['genre'].fillna('')

    # Mise en minuscule
    df['artist_name_lower'] = df['artist_name'].str.lower()
    df['track_name_lower'] = df['track_name'].str.lower()
    df['genre_lower'] = df['genre'].str.lower()

    # Listes de référence
    artistes_variete_fr = [
        "charles aznavour", "julien doré", "zaz", "vianney", "jean-jacques goldman",
        "edith piaf", "francis cabrel", "clara luciani", "coeur de pirate", "florent pagny",
        "michel sardou", "renaud", "alain souchon", "ben mazué", "patrick bruel"
    ]
    genres_cibles = ['chanson française', 'french pop', 'variété française', 'francophone']
    mots_cles = ['amour', 'coeur', 'liberté', 'vie', 'rêve', 'toi', 'moi', 'nous', 'ensemble', 'souvenir']

    # Calcul des scores
    df['score_artiste'] = df['artist_name_lower'].isin(artistes_variete_fr).astype(int)
    df['score_genre'] = df['genre_lower'].apply(lambda g: int(any(x in g for x in genres_cibles)))
    df['score_titre'] = df['track_name_lower'].apply(lambda t: int(any(m in t for m in mots_cles)))

    df['score_variete'] = df['score_artiste'] + df['score_genre'] + df['score_titre']

    return df
```

```python
# Reprise du script pour récupérer un maximum d'artistes francophones via MusicBrainz

import musicbrainzngs
import pandas as pd
import time

# Configuration de l'application
musicbrainzngs.set_useragent("MoodiFyApp", "1.0", "ton.email@example.com")

# Liste de pays francophones
pays_francophones = [
    "FR",  # France
    "BE",  # Belgique
    "CH",  # Suisse
    "CA",  # Canada
    "LU",  # Luxembourg
    "HT",  # Haïti
    "SN",  # Sénégal
    "CM",  # Cameroun
    "MG",  # Madagascar
    "CD",  # RDC
    "CI",  # Côte d’Ivoire
    "BF",  # Burkina Faso
    "TG",  # Togo
]

# Fonction pour récupérer les artistes par pays
def recuperer_artistes_par_pays(pays_list, total_max=2000, par_page=100):
    artistes = []
    for pays in pays_list:
        print(f"🔍 Récupération pour le pays : {pays}")
        for offset in range(0, total_max, par_page):
            try:
                resultats = musicbrainzngs.search_artists(
                    query=f"country:{pays}",
                    limit=par_page,
                    offset=offset
                )
                for artiste in resultats['artist-list']:
                    artistes.append({
                        'name': artiste.get('name', ''),
                        'country': artiste.get('country', ''),
                        'disambiguation': artiste.get('disambiguation', ''),
                        'score': artiste.get('ext:score', ''),
                        'tags': ', '.join([t['name'] for t in artiste.get('tag-list', [])]) if 'tag-list' in artiste else ''
                    })
                time.sleep(1.2)
            except Exception as e:
                print(f"❌ Erreur à l’offset {offset} pour {pays} : {e}")
                break
    return pd.DataFrame(artistes)

# Lancer la collecte
df_artistes_francophones = recuperer_artistes_par_pays(pays_francophones, total_max=300)

# Affichage du tableau final
import ace_tools as tools; tools.display_dataframe_to_user(name="Artistes francophones MusicBrainz", dataframe=df_artistes_francophones)

```

```python
# 3. Méthode IQR : comment ça marche ?
#Elle est basée sur les quartiles :
#Q1 (1er quartile) = 25 % des valeurs sont en dessous
#Q3 (3e quartile) = 75 % des valeurs sont en dessous

#IQR = Q3 - Q1 = l’intervalle au centre
#On considère comme "valeurs aberrantes" toutes celles hors de cet intervalle étendu :
#[ Q1 - 1.5 × IQR ; Q3 + 1.5 × IQR ]
```

```python
#appliquer la méthode IQR pour détecter les valeurs aberrantes
Q1 = df_spotify['valence'].quantile(0.25)
Q3 = df_spotify['valence'].quantile(0.75)
IQR = Q3 - Q1

df_spotify = df_spotify[~((df_spotify['valence'] < (Q1 - 1.5 * IQR)) | (df_spotify['valence'] > (Q3 + 1.5 * IQR)))]
# Vérification des valeurs restantes
print(df_spotify['valence'].describe())
```
