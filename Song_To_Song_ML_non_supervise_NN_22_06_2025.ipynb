```python
X_class = df.loc[:, ['valence', 'energy', 'danceability', 'acousticness', 'tempo', 'loudness', 'instrumentalness']]


# Affiche un titre pour indiquer le début de la section de classification avec NN
print("\n--- Classification (NN pour prédire 'genre') ---")


# Utiliser un ensemble cohérent de features pour trouver des voisins similaires
# Prenons les mêmes features que pour la classification (X_class)
# On crée un nouveau scaler et on l'entraîne sur l'ensemble de X_class
# Standardiser toutes ces features
scaler_nn = StandardScaler()
X_nn_scaled = scaler_nn.fit_transform(X_class) # Standardiser X_class



k = 5 # Nombre de voisins à trouver (y compris le point lui-même si k>0 et qu'il est dans les données)
# metric='euclidean' est la distance standard (ligne droite entre les points)
# algorithm='auto' choisit le meilleur algorithme en fonction des données
nn_model = NearestNeighbors(n_neighbors=k, algorithm='auto', metric='euclidean')
# .fit() indexe les données X_nn_scaled
nn_model.fit(X_nn_scaled) # Entraîner sur les données standardisées X_class


# Diviser les données (X_class et y_class)

# stratify=y_class est crucial ici pour gérer le déséquilibre potentiel des classes
X_train_c, X_test_c = train_test_split(X_class, test_size=0.2, random_state=42)

# Sélectionner les 3 premiers points de X_test_c (qui sont non standardisés)
sample_points_original = X_test_c[:3]

# Trouver les voisins pour quelques points de l'ensemble de test (X_test_c)
# Standardiser ces points d'abord avec le scaler_knn
X_test_c_scaled_knn = scaler_nn.transform(X_test_c) # Utiliser transform

# Sélectionner les 3 premiers points de X_test_c (qui sont non standardisés)
sample_points_scaled = X_test_c_scaled_knn[:3]
# .kneighbors() prend les points standardisés et retourne distances et indices
distances, indices = nn_model.kneighbors(sample_points_scaled)


print(f"\nRecherche des {k} plus proches voisins pour 3 morceaux (basé sur les features standardisées de X_class):")
for i in range(len(sample_points_scaled)):
    # Récupérer l'index original du point exemple dans le DataFrame df
    original_index = sample_points_original.index[i]
    print(f"\n--- Voisinage pour Morceaux Exemple {i+1} (Index original: {original_index}) ---")
    print(f" valence réelle: {df.loc[original_index, 'valence']:.2f}, energy réelle: {df.loc[original_index, 'energy']:.2f}, danceability réelle: {df.loc[original_index, 'danceability']:.2f}, tempo réelle: {df.loc[original_index, 'tempo']:.2f}, loudness réelle: {df.loc[original_index, 'loudness']:.2f}, instrumentalness réelle: {df.loc[original_index, 'instrumentalness']:.2f}  ")

    # Les 'indices' renvoyés par kneighbors sont les positions (0, 1, 2...) dans X_knn_scaled
    print(f"  Indices des voisins dans X_nn_scaled: {indices[i]}")
    # Les distances correspondantes
    print(f"  Distances euclidiennes aux voisins: {distances[i]}")

    # Pour afficher les informations des voisins, il faut retrouver leurs index originaux dans df.
    # Comme X_nn_scaled a été créé à partir de X_class dans le même ordre, on peut utiliser
    # les indices[i] pour sélectionner les lignes correspondantes dans X_class, puis récupérer leurs index originaux.
    neighbor_original_indices = X_class.iloc[indices[i]].index
    print(f"  Index originaux des voisins dans le DataFrame: {list(neighbor_original_indices)}")

    # Afficher quelques caractéristiques clés des voisins trouvés à partir du df original
    print("  Infos sur les voisins trouvés (depuis df original):")
    # Sélectionner les lignes des voisins dans df et quelques colonnes pertinentes
    neighbor_info = df.loc[neighbor_original_indices][['valence', 'energy', 'danceability', 'acousticness', 'tempo', 'loudness', 'instrumentalness']]
    print(neighbor_info)



# Fonction pour trouver les k voisins d'un morceaux par ID (avec signatures de type)

def find_neighbors_by_id(
    track_id: int,                      # L'index du morceaux cible dans le DataFrame original
    k: int,                            # Le nombre de voisins à rechercher
    nn_model: NearestNeighbors,        # Le modèle NearestNeighbors entraîné
    scaler: StandardScaler,            # Le scaler (StandardScaler) entraîné sur les features utilisées par nn_model
    features_df: pd.DataFrame,         # Le DataFrame contenant SEULEMENT les features utilisées pour entraîner nn_model (ex: X_class)
    original_df: pd.DataFrame          # Le DataFrame original complet pour afficher toutes les informations
) -> None:                             # La fonction ne retourne rien, elle affiche les résultats

    if track_id not in features_df.index:
        print(f"Erreur: L'ID de morceaux {track_id} n'existe pas dans les données utilisées.")
        return 

    # Récupérer les features du morceaux cible
    track_features = features_df.loc[[track_id]] # Garder le format DataFrame

    # Standardiser les features du morceaux cible avec le même scaler
    track_features_scaled = scaler.transform(track_features)

    # Trouver les k voisins (incluant potentiellement lui-même en premier)
    distances, indices = nn_model.kneighbors(track_features_scaled, n_neighbors=k)

    # Récupérer les index originaux des voisins trouvés
    neighbor_indices_in_features_df = indices[0] # indices est [[idx1, idx2,...]]

    # S'assurer que les indices sont valides pour .iloc
    valid_indices = [idx for idx in neighbor_indices_in_features_df if idx < len(features_df)]
    if len(valid_indices) < len(neighbor_indices_in_features_df):
        print("Avertissement: Certains indices de voisins étaient hors limites.")

    if not valid_indices:
        print("Aucun voisin valide trouvé.")
        return

    # Récupérer les index originaux des voisins trouvés dans le DataFrame original
    neighbor_original_indices = features_df.iloc[valid_indices].index

    print(f"\n--- Plus proches voisins pour le morceaux avec ID {track_id} ---")
    print("Morceaux Cible:")

    # Utiliser .loc pour être sûr avec l'index original
    print(original_df.loc[[track_id]])

    print(f"\n{len(neighbor_original_indices)} plus proches voisins trouvés (basé sur les features de X_class standardisées):")

    # Utiliser .loc avec la liste des index originaux des voisins
    neighbor_info = original_df.loc[neighbor_original_indices]
    print(neighbor_info)
    print("\nDistances euclidiennes correspondantes:")

    # Afficher seulement les distances pour les voisins valides trouvés
    print(distances[0][:len(valid_indices)])



# Choisir un ID de morceaux valide (par exemple, le premier index de l'ensemble de test X_test_c)

if not X_test_c.empty:
    example_id_to_find = X_test_c.index[0]
    num_neighbors_to_find = 6 # Chercher 5 voisins + lui-même (le plus proche voisin est souvent le point lui-même avec une distance de 0)

    # Appeler la fonction avec les objets nécessaires
    find_neighbors_by_id(
        track_id=example_id_to_find,
        k=num_neighbors_to_find,
        nn_model=nn_model,         # Le modèle NearestNeighbors entraîné
        scaler=scaler_nn,   # Le scaler entraîné sur tout X_class
        features_df=X_class,       # Le DataFrame des features utilisé pour entraîner nn_model
        original_df=df             # Le DataFrame original complet
    )
else:
    print("\nL'ensemble de test X_test_c est vide, impossible de choisir un example_id pour la fonction find_neighbors_by_id.")

# Tester avec un autre ID manuellement si vous le souhaitez
# find_neighbors_by_id(track_id=5018, k=5, nn_model=nn_model, scaler=scaler_nn_unsup, features_df=X_class, original_df=df)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
import warnings
import os
import platform
warnings.filterwarnings('ignore')

# CHARGEMENT AUTOMATIQUE DES DONNÉES SELON LE SYSTÈME D'EXPLOITATION
def charger_donnees():
    """
    Charge les données Spotify depuis le fichier CSV approprié selon l'OS
    
    Retourne:
    DataFrame : Données Spotify chargées
    """
    # Détection du système d'exploitation
    system = platform.system()
    
    # Définition des chemins selon l'OS
    if system == "Windows":
        # Chemin pour Windows
        chemin_fichier = r"C:\Users\beber\Documents\Cours\Wild_Code_School\Semaine_8_Visualisation_interactive_&_Pandas_approfondi\Projet_2\Projet_2_local\Datasets\df_concat_tagged.csv"
    else:
        # Chemin pour Mac/Linux
        chemin_fichier = "/Users/bertranddevulder/Documents/Cours/Wild_Code_School/Semaine_8_Visualisation_interactive_et_Pandas_approfondi/Projet_2/Projet_2_local/Datasets/df_concat_tagged.csv"
    
    try:
        # Tentative de chargement avec le chemin automatique
        df_spotify = pd.read_csv(chemin_fichier)
        print(f"Données chargées avec succès depuis: {chemin_fichier}")
        print(f"Forme du DataFrame: {df_spotify.shape}")
        return df_spotify
    except FileNotFoundError:
        # Si le fichier n'est pas trouvé, essayer des chemins alternatifs
        chemins_alternatifs = [
            "/Users/beber/Documents/Cours/Wild_Code_School/Semaine_8_Visualisation_interactive_&_Pandas_approfondi/Projet_2/Projet_2_local/Datasets/df_concat_tagged.csv",
            "./Datasets/df_concat_tagged.csv",
            "./df_concat_tagged.csv",
            "df_concat_tagged.csv"
        ]
        
        for chemin in chemins_alternatifs:
            try:
                df_spotify = pd.read_csv(chemin)
                print(f"Données chargées avec succès depuis: {chemin}")
                print(f"Forme du DataFrame: {df_spotify.shape}")
                return df_spotify
            except FileNotFoundError:
                continue
        
        # Si aucun chemin ne fonctionne, lever une erreur
        raise FileNotFoundError("Impossible de trouver le fichier df_concat_tagged.csv. Vérifiez le chemin.")

def test_clustering_configurations(df_spotify):
    """
    Fonction principale pour tester toutes les configurations de clustering
    
    Arguments:
    df_spotify : DataFrame contenant les données Spotify avec les features audio
    
    Retourne:
    dict : Dictionnaire contenant tous les résultats des tests
    """
    
    # Affichage des informations sur le DataFrame
    print(f"Colonnes disponibles: {list(df_spotify.columns)}")
    print(f"Nombre de lignes: {len(df_spotify)}")
    
    # Définition des différents cas de features à tester
    # Cas 1 : 7 features complètes incluant instrumentalness
    features_cas1 = ['valence', 'energy', 'danceability', 'acousticness', 
                     'tempo', 'loudness', 'instrumentalness']
    # Cas 2 : 6 features sans instrumentalness  
    features_cas2 = ['valence', 'energy', 'danceability', 'acousticness', 
                     'tempo', 'loudness']
    # Cas 3 : 5 features essentielles pour l'émotion musicale
    features_cas3 = ['valence', 'energy', 'danceability', 'acousticness', 'tempo']
    
    cases_features = {
        'cas1': features_cas1,
        'cas2': features_cas2, 
        'cas3': features_cas3
    }
    
    # Vérification de la disponibilité des features dans le DataFrame
    print("\nVérification des features disponibles:")
    for cas_nom, features in cases_features.items():
        features_manquantes = [f for f in features if f not in df_spotify.columns]
        if features_manquantes:
            print(f"ATTENTION - {cas_nom}: Features manquantes: {features_manquantes}")
        else:
            print(f"OK - {cas_nom}: Toutes les features disponibles")
    
    # Paramètres à tester pour le clustering
    k_values = [2, 3, 4, 5, 7, 10, 15, 20, 25]  # Suppression de k=1 car impossible pour le clustering
    distance_metrics = ['euclidean', 'manhattan']  # Métriques de distance
    algorithms = ['auto', 'ball_tree', 'kd_tree']  # Algorithmes pour NearestNeighbors
    scalers = {
        'StandardScaler': StandardScaler(),  # Standardisation avec moyenne=0, écart-type=1
        'MinMaxScaler': MinMaxScaler(),      # Normalisation entre 0 et 1
        'RobustScaler': RobustScaler()       # Normalisation robuste aux outliers
    }
    
    # Paramètres DBSCAN - eps et min_samples sont cruciaux pour la performance
    eps_values = [0.3, 0.5, 0.7, 1.0, 1.5]  # Rayon de voisinage
    min_samples_values = [3, 5, 7, 10]       # Nombre minimum de points pour former un cluster
    
    resultats = []  # Stockage de tous les résultats
    
    print("\nDébut des tests de clustering...")
    
    # Boucle principale pour tester toutes les combinaisons
    for cas_nom, features in cases_features.items():
        print(f"\nTest du {cas_nom} avec {len(features)} features")
        
        # Vérification que toutes les features existent dans le DataFrame
        features_disponibles = [f for f in features if f in df_spotify.columns]
        if len(features_disponibles) != len(features):
            print(f"ERREUR: Certaines features manquantes pour {cas_nom}")
            continue
            
        # Extraction des données pour ce cas
        X = df_spotify[features_disponibles].copy()
        
        # Nettoyage des données - suppression des valeurs manquantes et infinies
        print(f"Données avant nettoyage: {len(X)} lignes")
        X = X.dropna()
        X = X.replace([np.inf, -np.inf], np.nan).dropna()
        print(f"Données après nettoyage: {len(X)} lignes")
        
        if len(X) < 100:  # Vérification d'un minimum de données
            print(f"ERREUR: Pas assez de données pour {cas_nom} ({len(X)} lignes)")
            continue
            
        for scaler_nom, scaler in scalers.items():
            print(f"  Test avec {scaler_nom}...")
            
            try:
                # Application du prétraitement - scaling des données
                scaler_fitted = scaler.fit(X)
                X_scaled = scaler_fitted.transform(X)
                
                # Test des modèles K-means
                for k in k_values:
                    if k > len(X):  # K ne peut pas être supérieur au nombre d'échantillons
                        continue
                        
                    try:
                        # Initialisation et entraînement du modèle K-means
                        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
                        kmeans_labels = kmeans.fit_predict(X_scaled)
                        
                        # Vérification qu'il y a effectivement k clusters différents
                        if len(np.unique(kmeans_labels)) < 2:
                            continue
                            
                        # Calcul des métriques d'évaluation
                        # Silhouette Score : mesure la cohésion intra-cluster et séparation inter-cluster
                        # Valeurs entre -1 et 1, plus c'est proche de 1, mieux c'est
                        sil_score = silhouette_score(X_scaled, kmeans_labels)
                        
                        # Davies-Bouldin Score : mesure la similarité intra-cluster vs distance inter-cluster  
                        # Plus la valeur est faible, meilleur est le clustering
                        db_score = davies_bouldin_score(X_scaled, kmeans_labels)
                        
                        # Calinski-Harabasz Score : ratio de la dispersion inter-cluster sur intra-cluster
                        # Plus la valeur est élevée, meilleur est le clustering
                        ch_score = calinski_harabasz_score(X_scaled, kmeans_labels)
                        
                        # Stockage des résultats K-means
                        resultats.append({
                            'cas': cas_nom,
                            'features_count': len(features_disponibles),
                            'scaler': scaler_nom,
                            'modele': 'KMeans',
                            'k_clusters': k,
                            'silhouette_score': sil_score,
                            'davies_bouldin_score': db_score,
                            'calinski_harabasz_score': ch_score,
                            'n_samples': len(X_scaled),
                            'features_list': ', '.join(features_disponibles)
                        })
                        
                    except Exception as e:
                        print(f"    Erreur K-means k={k}: {str(e)}")
                        continue
                
                # Test des modèles DBSCAN
                for eps in eps_values:
                    for min_samples in min_samples_values:
                        try:
                            # Initialisation et entraînement du modèle DBSCAN
                            # DBSCAN identifie les clusters de densité variable et détecte le bruit
                            dbscan = DBSCAN(eps=eps, min_samples=min_samples)
                            dbscan_labels = dbscan.fit_predict(X_scaled)
                            
                            # Vérification qu'il y a des clusters (pas que du bruit)
                            n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
                            if n_clusters < 2:
                                continue
                                
                            # Exclusion du bruit (label -1) pour le calcul des métriques
                            mask = dbscan_labels != -1
                            if np.sum(mask) < 10:  # Au moins 10 points non-bruit
                                continue
                                
                            X_no_noise = X_scaled[mask]
                            labels_no_noise = dbscan_labels[mask]
                            
                            # Calcul des métriques d'évaluation
                            sil_score = silhouette_score(X_no_noise, labels_no_noise)
                            db_score = davies_bouldin_score(X_no_noise, labels_no_noise)
                            ch_score = calinski_harabasz_score(X_no_noise, labels_no_noise)
                            
                            # Stockage des résultats DBSCAN
                            resultats.append({
                                'cas': cas_nom,
                                'features_count': len(features_disponibles),
                                'scaler': scaler_nom,
                                'modele': 'DBSCAN',
                                'eps': eps,
                                'min_samples': min_samples,
                                'n_clusters': n_clusters,
                                'noise_points': np.sum(dbscan_labels == -1),
                                'silhouette_score': sil_score,
                                'davies_bouldin_score': db_score,
                                'calinski_harabasz_score': ch_score,
                                'n_samples': len(X_scaled),
                                'features_list': ', '.join(features_disponibles)
                            })
                            
                        except Exception as e:
                            print(f"    Erreur DBSCAN eps={eps}, min_samples={min_samples}: {str(e)}")
                            continue
                            
            except Exception as e:
                print(f"  Erreur preprocessing {scaler_nom}: {str(e)}")
                continue
    
    print(f"\nTests terminés. Nombre total de configurations testées: {len(resultats)}")
    return resultats

def analyser_resultats(resultats):
    """
    Analyse des résultats et génération des Top 5
    
    Arguments:
    resultats : Liste des dictionnaires contenant tous les résultats
    
    Retourne:
    dict : Analyses et Top 5 organisés
    """
    
    if not resultats:
        print("Aucun résultat à analyser")
        return {}
    
    df_resultats = pd.DataFrame(resultats)
    
    # Séparation des résultats par modèle
    kmeans_results = df_resultats[df_resultats['modele'] == 'KMeans'].copy()
    dbscan_results = df_resultats[df_resultats['modele'] == 'DBSCAN'].copy()
    
    analyses = {}
    
    # TOP 5 K-MEANS
    print("\n" + "="*80)
    print("TOP 5 MEILLEURS RÉSULTATS K-MEANS")
    print("="*80)
    
    if not kmeans_results.empty:
        # Tri par silhouette_score décroissant (meilleur = plus élevé)
        top5_kmeans_sil = kmeans_results.nlargest(5, 'silhouette_score')
        print("\nTOP 5 - Silhouette Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_kmeans_sil.iterrows(), 1):
            print(f"{idx}. Score: {row['silhouette_score']:.4f} | "
                  f"Cas: {row['cas']} | Scaler: {row['scaler']} | "
                  f"K-clusters: {row['k_clusters']} | Samples: {row['n_samples']}")
        
        # Tri par davies_bouldin_score croissant (meilleur = plus faible)
        top5_kmeans_db = kmeans_results.nsmallest(5, 'davies_bouldin_score')
        print("\nTOP 5 - Davies-Bouldin Score (meilleur = plus faible):")
        for idx, (i, row) in enumerate(top5_kmeans_db.iterrows(), 1):
            print(f"{idx}. Score: {row['davies_bouldin_score']:.4f} | "
                  f"Cas: {row['cas']} | Scaler: {row['scaler']} | "
                  f"K-clusters: {row['k_clusters']} | Samples: {row['n_samples']}")
        
        # Tri par calinski_harabasz_score décroissant (meilleur = plus élevé)
        top5_kmeans_ch = kmeans_results.nlargest(5, 'calinski_harabasz_score')
        print("\nTOP 5 - Calinski-Harabasz Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_kmeans_ch.iterrows(), 1):
            print(f"{idx}. Score: {row['calinski_harabasz_score']:.4f} | "
                  f"Cas: {row['cas']} | Scaler: {row['scaler']} | "
                  f"K-clusters: {row['k_clusters']} | Samples: {row['n_samples']}")
    
    # TOP 5 DBSCAN
    print("\n" + "="*80)
    print("TOP 5 MEILLEURS RÉSULTATS DBSCAN")
    print("="*80)
    
    if not dbscan_results.empty:
        # Tri par silhouette_score décroissant
        top5_dbscan_sil = dbscan_results.nlargest(5, 'silhouette_score')
        print("\nTOP 5 - Silhouette Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_dbscan_sil.iterrows(), 1):
            print(f"{idx}. Score: {row['silhouette_score']:.4f} | "
                  f"Cas: {row['cas']} | Scaler: {row['scaler']} | "
                  f"EPS: {row['eps']} | Min-samples: {row['min_samples']} | "
                  f"Clusters: {row['n_clusters']} | Bruit: {row['noise_points']}")
        
        # Tri par davies_bouldin_score croissant
        top5_dbscan_db = dbscan_results.nsmallest(5, 'davies_bouldin_score')
        print("\nTOP 5 - Davies-Bouldin Score (meilleur = plus faible):")
        for idx, (i, row) in enumerate(top5_dbscan_db.iterrows(), 1):
            print(f"{idx}. Score: {row['davies_bouldin_score']:.4f} | "
                  f"Cas: {row['cas']} | Scaler: {row['scaler']} | "
                  f"EPS: {row['eps']} | Min-samples: {row['min_samples']} | "
                  f"Clusters: {row['n_clusters']} | Bruit: {row['noise_points']}")
        
        # Tri par calinski_harabasz_score décroissant
        top5_dbscan_ch = dbscan_results.nlargest(5, 'calinski_harabasz_score')
        print("\nTOP 5 - Calinski-Harabasz Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_dbscan_ch.iterrows(), 1):
            print(f"{idx}. Score: {row['calinski_harabasz_score']:.4f} | "
                  f"Cas: {row['cas']} | Scaler: {row['scaler']} | "
                  f"EPS: {row['eps']} | Min-samples: {row['min_samples']} | "
                  f"Clusters: {row['n_clusters']} | Bruit: {row['noise_points']}")
    
    # TOP 5 GLOBAL - Combinaison des meilleurs modèles
    print("\n" + "="*80)
    print("TOP 5 MEILLEURS MODÈLES GLOBAUX")
    print("="*80)
    
    # Normalisation des scores pour comparaison inter-modèles
    # Création d'un score composite basé sur les 3 métriques
    df_resultats_copy = df_resultats.copy()
    
    # Normalisation min-max des scores entre 0 et 1
    df_resultats_copy['sil_norm'] = (df_resultats_copy['silhouette_score'] - df_resultats_copy['silhouette_score'].min()) / (df_resultats_copy['silhouette_score'].max() - df_resultats_copy['silhouette_score'].min())
    df_resultats_copy['db_norm'] = 1 - ((df_resultats_copy['davies_bouldin_score'] - df_resultats_copy['davies_bouldin_score'].min()) / (df_resultats_copy['davies_bouldin_score'].max() - df_resultats_copy['davies_bouldin_score'].min()))
    df_resultats_copy['ch_norm'] = (df_resultats_copy['calinski_harabasz_score'] - df_resultats_copy['calinski_harabasz_score'].min()) / (df_resultats_copy['calinski_harabasz_score'].max() - df_resultats_copy['calinski_harabasz_score'].min())
    
    # Score composite : moyenne pondérée des 3 métriques normalisées
    df_resultats_copy['score_composite'] = (df_resultats_copy['sil_norm'] * 0.4 + 
                                           df_resultats_copy['db_norm'] * 0.3 + 
                                           df_resultats_copy['ch_norm'] * 0.3)
    
    top5_global = df_resultats_copy.nlargest(5, 'score_composite')
    print("\nTOP 5 - Score Composite (silhouette 40% + davies-bouldin 30% + calinski-harabasz 30%):")
    for idx, (i, row) in enumerate(top5_global.iterrows(), 1):
        config_details = ""
        if row['modele'] == 'KMeans':
            config_details = f"K-clusters: {row['k_clusters']}"
        else:
            config_details = f"EPS: {row['eps']} | Min-samples: {row['min_samples']} | Clusters: {row['n_clusters']}"
        
        print(f"{idx}. Score: {row['score_composite']:.4f} | "
              f"Modèle: {row['modele']} | Cas: {row['cas']} | "
              f"Scaler: {row['scaler']} | {config_details}")
    
    # ANALYSE GLOBALE DES PARAMÈTRES INFLUENTS
    print("\n" + "="*80)
    print("ANALYSE GLOBALE DES PARAMÈTRES INFLUENTS")
    print("="*80)
    
    # Analyse par cas de features
    print("\nINFLUENCE DU NOMBRE DE FEATURES:")
    for cas in df_resultats['cas'].unique():
        subset = df_resultats[df_resultats['cas'] == cas]
        avg_sil = subset['silhouette_score'].mean()
        std_sil = subset['silhouette_score'].std()
        print(f"- {cas} ({subset.iloc[0]['features_count']} features): "
              f"Silhouette moyen = {avg_sil:.4f} ± {std_sil:.4f}")
    
    # Analyse par scaler
    print("\nINFLUENCE DU TYPE DE SCALING:")
    for scaler in df_resultats['scaler'].unique():
        subset = df_resultats[df_resultats['scaler'] == scaler]
        avg_sil = subset['silhouette_score'].mean()
        std_sil = subset['silhouette_score'].std()
        print(f"- {scaler}: Silhouette moyen = {avg_sil:.4f} ± {std_sil:.4f}")
    
    # Analyse par modèle
    print("\nPERFORMANCE COMPARATIVE DES MODÈLES:")
    for modele in df_resultats['modele'].unique():
        subset = df_resultats[df_resultats['modele'] == modele]
        avg_sil = subset['silhouette_score'].mean()
        avg_db = subset['davies_bouldin_score'].mean()
        avg_ch = subset['calinski_harabasz_score'].mean()
        print(f"- {modele}: Sil={avg_sil:.4f}, DB={avg_db:.4f}, CH={avg_ch:.2f}")
    
    print("\n" + "="*80)
    print("RECOMMANDATIONS POUR LE DÉPLOIEMENT")
    print("="*80)
    
    best_model = top5_global.iloc[0]
    print(f"\nMEILLEUR MODÈLE RECOMMANDÉ:")
    print(f"- Modèle: {best_model['modele']}")
    print(f"- Cas de features: {best_model['cas']}")
    print(f"- Préprocessing: {best_model['scaler']}")
    print(f"- Features utilisées: {best_model['features_list']}")
    
    if best_model['modele'] == 'KMeans':
        print(f"- Nombre de clusters: {best_model['k_clusters']}")
        print(f"\nCODE POUR LE DÉPLOIEMENT:")
        print(f"# Chargement et preprocessing")
        print(f"from sklearn.preprocessing import {best_model['scaler']}")
        print(f"from sklearn.cluster import KMeans")
        print(f"")
        print(f"# Sélection des features")
        print(f"features = {best_model['features_list'].split(', ')}")
        print(f"X = df[features]")
        print(f"")
        print(f"# Preprocessing")
        print(f"scaler = {best_model['scaler']}()")
        print(f"X_scaled = scaler.fit_transform(X)")
        print(f"")
        print(f"# Modèle")
        print(f"model = KMeans(n_clusters={best_model['k_clusters']}, random_state=42)")
        print(f"clusters = model.fit_predict(X_scaled)")
        print(f"")
        print(f"# Score de qualité attendu: {best_model['silhouette_score']:.4f}")
    else:
        print(f"- EPS: {best_model['eps']}")
        print(f"- Min samples: {best_model['min_samples']}")
        print(f"- Clusters détectés: {best_model['n_clusters']}")
        print(f"\nCODE POUR LE DÉPLOIEMENT:")
        print(f"# Chargement et preprocessing")
        print(f"from sklearn.preprocessing import {best_model['scaler']}")
        print(f"from sklearn.cluster import DBSCAN")
        print(f"")
        print(f"# Sélection des features")
        print(f"features = {best_model['features_list'].split(', ')}")
        print(f"X = df[features]")
        print(f"")
        print(f"# Preprocessing")
        print(f"scaler = {best_model['scaler']}()")
        print(f"X_scaled = scaler.fit_transform(X)")
        print(f"")
        print(f"# Modèle")
        print(f"model = DBSCAN(eps={best_model['eps']}, min_samples={best_model['min_samples']})")
        print(f"clusters = model.fit_predict(X_scaled)")
        print(f"")
        print(f"# Score de qualité attendu: {best_model['silhouette_score']:.4f}")
    
    # EXPLICATION DES MÉTRIQUES
    print("\n" + "="*80)
    print("EXPLICATION DES MÉTRIQUES D'ÉVALUATION")
    print("="*80)
    
    print("\n1. SILHOUETTE SCORE:")
    print("   - Mesure la cohésion intra-cluster vs séparation inter-cluster")
    print("   - Valeurs entre -1 et 1, optimal proche de 1")
    print("   - Indique si les points sont bien assignés à leur cluster")
    print("   - Pénalise les clusters trop proches ou mal séparés")
    
    print("\n2. DAVIES-BOULDIN SCORE:")
    print("   - Mesure la similarité moyenne entre chaque cluster et son plus similaire")
    print("   - Plus la valeur est faible, meilleur est le clustering")
    print("   - Évalue la compacité intra-cluster et la séparation inter-cluster")
    print("   - Particulièrement sensible aux clusters de formes sphériques")
    
    print("\n3. CALINSKI-HARABASZ SCORE:")
    print("   - Ratio de la variance inter-cluster sur la variance intra-cluster")
    print("   - Plus la valeur est élevée, meilleur est le clustering")
    print("   - Favorise les clusters compacts et bien séparés")
    print("   - Aussi appelé Variance Ratio Criterion")
    
    print("\n" + "="*80)
    print("CONSEILS D'UTILISATION EN PRODUCTION")
    print("="*80)
    
    print("\n1. PRÉPARATION DES DONNÉES:")
    print("   - Appliquer le même preprocessing que lors de l'entraînement")
    print("   - Vérifier la qualité des données d'entrée")
    print("   - Gérer les valeurs manquantes de manière cohérente")
    
    print("\n2. MONITORING:")
    print("   - Surveiller la distribution des clusters au fil du temps")
    print("   - Recalculer périodiquement les métriques sur de nouvelles données")
    print("   - Réentraîner le modèle si les performances se dégradent")
    
    print("\n3. INTERPRÉTATION:")
    print("   - Analyser les caractéristiques moyennes de chaque cluster")
    print("   - Créer des profils musicaux basés sur les clusters")
    print("   - Utiliser pour la recommandation ou la classification automatique")
    
    # Sauvegarde des résultats
    try:
        df_resultats.to_csv('resultats_clustering_complets.csv', index=False)
        top5_global.to_csv('top5_meilleurs_modeles.csv', index=False)
        print(f"\nRésultats sauvegardés dans:")
        print(f"- resultats_clustering_complets.csv")
        print(f"- top5_meilleurs_modeles.csv")
    except Exception as e:
        print(f"Erreur lors de la sauvegarde: {e}")
    
    return {
        'tous_resultats': df_resultats,
        'top5_global': top5_global,
        'meilleur_modele': best_model,
        'analyses': analyses
    }

# FONCTION PRINCIPALE D'EXÉCUTION
def executer_tests_complets():
    """
    Fonction principale pour exécuter tous les tests de clustering
    
    Cette fonction orchestre l'ensemble du processus de test:
    1. Charge les données depuis le fichier CSV
    2. Teste toutes les configurations possibles
    3. Analyse les résultats 
    4. Génère les recommandations
    
    Retourne:
    dict : Résultats complets de l'analyse
    """
    
    print("DÉBUT DE L'ANALYSE COMPLÈTE DES MODÈLES DE CLUSTERING")
    print("="*80)
    print(f"Système d'exploitation détecté: {platform.system()}")
    
    try:
        # Chargement des données
        df_spotify = charger_donnees()
        
        # Exécution de tous les tests
        resultats = test_clustering_configurations(df_spotify)
        
        if not resultats:
            print("ERREUR: Aucun résultat généré")
            return None
        
        print(f"\nNOMBRE TOTAL DE CONFIGURATIONS TESTÉES: {len(resultats)}")
        
        # Analyse des résultats
        analyses = analyser_resultats(resultats)
        
        print("\nANALYSE TERMINÉE AVEC SUCCÈS!")
        
        return analyses
        
    except Exception as e:
        print(f"ERREUR LORS DE L'EXÉCUTION: {e}")
        return None

# EXÉCUTION AUTOMATIQUE
if __name__ == "__main__":
    resultats_finaux = executer_tests_complets()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.pipeline import Pipeline
import warnings
import os
import platform
# Suppression de l'import fpdf qui cause l'erreur
# from fpdf import FPDF
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

def charger_donnees():
    """
    Charge les données Spotify depuis le fichier CSV approprié selon l'OS
    
    Retourne:
    DataFrame : Données Spotify chargées
    """
    # Détection du système d'exploitation
    system = platform.system()
    
    # Définition des chemins selon l'OS
    if system == "Windows":
        chemin_fichier = r"C:\Users\beber\Documents\Cours\Wild_Code_School\Semaine_8_Visualisation_interactive_&_Pandas_approfondi\Projet_2\Projet_2_local\Datasets\df_concat_tagged.csv"
    else:
        chemin_fichier = "/Users/bertranddevulder/Documents/Cours/Wild_Code_School/Semaine_8_Visualisation_interactive_et_Pandas_approfondi/Projet_2/Projet_2_local/Datasets/df_concat_tagged.csv"
    
    try:
        # Tentative de chargement avec le chemin automatique
        df_spotify = pd.read_csv(chemin_fichier)
        print(f"Données chargées avec succès depuis: {chemin_fichier}")
        print(f"Forme du DataFrame: {df_spotify.shape}")
        return df_spotify
    except FileNotFoundError:
        # Si le fichier n'est pas trouvé, essayer des chemins alternatifs
        chemins_alternatifs = [
            "/Users/beber/Documents/Cours/Wild_Code_School/Semaine_8_Visualisation_interactive_&_Pandas_approfondi/Projet_2/Projet_2_local/Datasets/df_concat_tagged.csv",
            "./Datasets/df_concat_tagged.csv",
            "./df_concat_tagged.csv",
            "df_concat_tagged.csv"
        ]
        
        for chemin in chemins_alternatifs:
            try:
                df_spotify = pd.read_csv(chemin)
                print(f"Données chargées avec succès depuis: {chemin}")
                print(f"Forme du DataFrame: {df_spotify.shape}")
                return df_spotify
            except FileNotFoundError:
                continue
        
        raise FileNotFoundError("Impossible de trouver le fichier df_concat_tagged.csv. Vérifiez le chemin.")

def test_configurations_ml_complet(df_spotify):
    """
    Teste toutes les configurations possibles de modèles ML non supervisés
    
    Arguments:
    df_spotify : DataFrame contenant les données Spotify avec les features audio
    
    Retourne:
    list : Liste des dictionnaires contenant tous les résultats des tests
    """
    
    # Affichage des informations sur le DataFrame
    print(f"Colonnes disponibles: {list(df_spotify.columns)}")
    print(f"Nombre de lignes: {len(df_spotify)}")
    
    # Définition des cas de features à tester selon les spécifications
    # Cas 1 : 7 features complètes incluant instrumentalness
    features_cas1 = ['valence', 'energy', 'danceability', 'acousticness', 
                     'tempo', 'loudness', 'instrumentalness']
    # Cas 2 : 6 features sans instrumentalness  
    features_cas2 = ['valence', 'energy', 'danceability', 'acousticness', 
                     'tempo', 'loudness']
    # Cas 3 : 5 features essentielles pour l'émotion musicale
    features_cas3 = ['valence', 'energy', 'danceability', 'acousticness', 'tempo']
    
    cases_features = {
        'cas1': features_cas1,
        'cas2': features_cas2, 
        'cas3': features_cas3
    }
    
    # Vérification de la disponibilité des features dans le DataFrame
    print("\nVérification des features disponibles:")
    for cas_nom, features in cases_features.items():
        features_manquantes = [f for f in features if f not in df_spotify.columns]
        if features_manquantes:
            print(f"ATTENTION - {cas_nom}: Features manquantes: {features_manquantes}")
        else:
            print(f"OK - {cas_nom}: Toutes les features disponibles")
    
    # Paramètres NearestNeighbors selon les spécifications
    k_values = [1, 2, 3, 4, 5, 7, 10, 15, 20, 25]  # Valeurs k pour NN
    distance_metrics = ['euclidean', 'manhattan']     # Métriques de distance
    algorithms = ['auto', 'ball_tree', 'kd_tree']    # Algorithmes NN
    
    # Scalers pour le préprocessing selon les spécifications
    scalers = {
        'StandardScaler': StandardScaler(),  # Standardisation avec moyenne=0, écart-type=1
        'MinMaxScaler': MinMaxScaler(),      # Normalisation entre 0 et 1
        'RobustScaler': RobustScaler()       # Normalisation robuste aux outliers
    }
    
    # Paramètres pour les modèles de clustering
    kmeans_k_values = [2, 3, 4, 5, 7, 10, 15, 20, 25]  # Nombre de clusters K-means
    eps_values = [0.3, 0.5, 0.7, 1.0, 1.5]             # Rayon de voisinage DBSCAN
    min_samples_values = [3, 5, 7, 10]                  # Nombre minimum de points DBSCAN
    
    resultats = []  # Stockage de tous les résultats
    
    print("\nDébut des tests exhaustifs de clustering...")
    
    # Boucle principale pour tester toutes les combinaisons possibles
    for cas_nom, features in cases_features.items():
        print(f"\nTest du {cas_nom} avec {len(features)} features")
        
        # Vérification que toutes les features existent dans le DataFrame
        features_disponibles = [f for f in features if f in df_spotify.columns]
        if len(features_disponibles) != len(features):
            print(f"ERREUR: Certaines features manquantes pour {cas_nom}")
            continue
            
        # Extraction des données pour ce cas
        X = df_spotify[features_disponibles].copy()
        
        # Nettoyage des données - suppression des valeurs manquantes et infinies
        print(f"Données avant nettoyage: {len(X)} lignes")
        X = X.dropna()
        X = X.replace([np.inf, -np.inf], np.nan).dropna()
        print(f"Données après nettoyage: {len(X)} lignes")
        
        if len(X) < 100:  # Vérification d'un minimum de données
            print(f"ERREUR: Pas assez de données pour {cas_nom} ({len(X)} lignes)")
            continue
            
        for scaler_nom, scaler in scalers.items():
            print(f"  Test avec {scaler_nom}...")
            
            try:
                # Application du prétraitement - scaling des données
                scaler_fitted = scaler.fit(X)
                X_scaled = scaler_fitted.transform(X)
                
                # Test de tous les paramètres NearestNeighbors
                for k_nn in k_values:
                    # K ne peut pas être supérieur au nombre d'échantillons
                    if k_nn > len(X_scaled):
                        continue
                        
                    for distance_metric in distance_metrics:
                        for algorithm in algorithms:
                            
                            # Test de compatibilité NearestNeighbors
                            try:
                                # Initialisation et test du modèle NearestNeighbors
                                nn = NearestNeighbors(
                                    n_neighbors=k_nn, 
                                    metric=distance_metric, 
                                    algorithm=algorithm
                                )
                                nn.fit(X_scaled)
                                
                            except Exception as e:
                                # Certaines combinaisons métrique/algorithme sont incompatibles
                                print(f"    Combinaison incompatible NN: k={k_nn}, metric={distance_metric}, algo={algorithm}")
                                continue
                            
                            # Test des modèles K-means avec cette configuration NN
                            for k_kmeans in kmeans_k_values:
                                if k_kmeans > len(X_scaled):  # K ne peut pas être supérieur au nombre d'échantillons
                                    continue
                                    
                                try:
                                    # Initialisation et entraînement du modèle K-means
                                    kmeans = KMeans(n_clusters=k_kmeans, random_state=42, n_init=10)
                                    kmeans_labels = kmeans.fit_predict(X_scaled)
                                    
                                    # Vérification qu'il y a effectivement des clusters différents
                                    if len(np.unique(kmeans_labels)) < 2:
                                        continue
                                        
                                    # Calcul des métriques d'évaluation pour K-means
                                    # Silhouette Score : mesure la cohésion intra-cluster et séparation inter-cluster
                                    # Valeurs entre -1 et 1, plus c'est proche de 1, mieux c'est
                                    sil_score = silhouette_score(X_scaled, kmeans_labels)
                                    
                                    # Davies-Bouldin Score : mesure la similarité intra-cluster vs distance inter-cluster  
                                    # Plus la valeur est faible, meilleur est le clustering
                                    db_score = davies_bouldin_score(X_scaled, kmeans_labels)
                                    
                                    # Calinski-Harabasz Score : ratio de la dispersion inter-cluster sur intra-cluster
                                    # Plus la valeur est élevée, meilleur est le clustering
                                    ch_score = calinski_harabasz_score(X_scaled, kmeans_labels)
                                    
                                    # Stockage des résultats K-means avec configuration NN complète
                                    resultats.append({
                                        'cas': cas_nom,
                                        'features_count': len(features_disponibles),
                                        'features_list': ', '.join(features_disponibles),
                                        'scaler': scaler_nom,
                                        'nn_k': k_nn,
                                        'nn_metric': distance_metric,
                                        'nn_algorithm': algorithm,
                                        'modele_clustering': 'KMeans',
                                        'k_clusters': k_kmeans,
                                        'eps': None,
                                        'min_samples': None,
                                        'silhouette_score': sil_score,
                                        'davies_bouldin_score': db_score,
                                        'calinski_harabasz_score': ch_score,
                                        'n_samples': len(X_scaled)
                                    })
                                    
                                except Exception as e:
                                    print(f"    Erreur K-means k={k_kmeans}: {str(e)}")
                                    continue
                            
                            # Test des modèles DBSCAN avec cette configuration NN
                            for eps in eps_values:
                                for min_samples in min_samples_values:
                                    try:
                                        # Initialisation et entraînement du modèle DBSCAN
                                        # DBSCAN identifie les clusters de densité variable et détecte le bruit
                                        dbscan = DBSCAN(eps=eps, min_samples=min_samples)
                                        dbscan_labels = dbscan.fit_predict(X_scaled)
                                        
                                        # Vérification qu'il y a des clusters (pas que du bruit)
                                        n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
                                        if n_clusters < 2:
                                            continue
                                            
                                        # Exclusion du bruit (label -1) pour le calcul des métriques
                                        mask = dbscan_labels != -1
                                        if np.sum(mask) < 10:  # Au moins 10 points non-bruit
                                            continue
                                            
                                        X_no_noise = X_scaled[mask]
                                        labels_no_noise = dbscan_labels[mask]
                                        
                                        # Calcul des métriques d'évaluation pour DBSCAN
                                        sil_score = silhouette_score(X_no_noise, labels_no_noise)
                                        db_score = davies_bouldin_score(X_no_noise, labels_no_noise)
                                        ch_score = calinski_harabasz_score(X_no_noise, labels_no_noise)
                                        
                                        # Stockage des résultats DBSCAN avec configuration NN complète
                                        resultats.append({
                                            'cas': cas_nom,
                                            'features_count': len(features_disponibles),
                                            'features_list': ', '.join(features_disponibles),
                                            'scaler': scaler_nom,
                                            'nn_k': k_nn,
                                            'nn_metric': distance_metric,
                                            'nn_algorithm': algorithm,
                                            'modele_clustering': 'DBSCAN',
                                            'k_clusters': None,
                                            'eps': eps,
                                            'min_samples': min_samples,
                                            'n_clusters_detectes': n_clusters,
                                            'noise_points': np.sum(dbscan_labels == -1),
                                            'silhouette_score': sil_score,
                                            'davies_bouldin_score': db_score,
                                            'calinski_harabasz_score': ch_score,
                                            'n_samples': len(X_scaled)
                                        })
                                        
                                    except Exception as e:
                                        print(f"    Erreur DBSCAN eps={eps}, min_samples={min_samples}: {str(e)}")
                                        continue
                            
            except Exception as e:
                print(f"  Erreur preprocessing {scaler_nom}: {str(e)}")
                continue
    
    print(f"\nTests terminés. Nombre total de configurations testées: {len(resultats)}")
    return resultats

def analyser_et_classer_resultats(resultats):
    """
    Analyse les résultats et génère les Top 5 demandés avec analyses complètes
    
    Arguments:
    resultats : Liste des dictionnaires contenant tous les résultats
    
    Retourne:
    dict : Analyses complètes et Top 5 organisés
    """
    
    if not resultats:
        print("Aucun résultat à analyser")
        return {}
    
    df_resultats = pd.DataFrame(resultats)
    
    # Séparation des résultats par modèle de clustering
    kmeans_results = df_resultats[df_resultats['modele_clustering'] == 'KMeans'].copy()
    dbscan_results = df_resultats[df_resultats['modele_clustering'] == 'DBSCAN'].copy()
    
    print("\n" + "="*100)
    print("TOP 5 MEILLEURS RÉSULTATS K-MEANS")
    print("="*100)
    
    if not kmeans_results.empty:
        # Tri par silhouette_score décroissant (meilleur = plus élevé)
        top5_kmeans_sil = kmeans_results.nlargest(5, 'silhouette_score')
        print("\nTOP 5 - Silhouette Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_kmeans_sil.iterrows(), 1):
            print(f"{idx}. Score Silhouette: {row['silhouette_score']:.4f}")
            print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
            print(f"   Configuration K-Means: k_clusters={row['k_clusters']}")
            print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
            print(f"   Échantillons: {row['n_samples']}")
            print()
        
        # Tri par davies_bouldin_score croissant (meilleur = plus faible)
        top5_kmeans_db = kmeans_results.nsmallest(5, 'davies_bouldin_score')
        print("TOP 5 - Davies-Bouldin Score (meilleur = plus faible):")
        for idx, (i, row) in enumerate(top5_kmeans_db.iterrows(), 1):
            print(f"{idx}. Score Davies-Bouldin: {row['davies_bouldin_score']:.4f}")
            print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
            print(f"   Configuration K-Means: k_clusters={row['k_clusters']}")
            print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
            print(f"   Échantillons: {row['n_samples']}")
            print()
        
        # Tri par calinski_harabasz_score décroissant (meilleur = plus élevé)
        top5_kmeans_ch = kmeans_results.nlargest(5, 'calinski_harabasz_score')
        print("TOP 5 - Calinski-Harabasz Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_kmeans_ch.iterrows(), 1):
            print(f"{idx}. Score Calinski-Harabasz: {row['calinski_harabasz_score']:.2f}")
            print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
            print(f"   Configuration K-Means: k_clusters={row['k_clusters']}")
            print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
            print(f"   Échantillons: {row['n_samples']}")
            print()
    
    print("\n" + "="*100)
    print("TOP 5 MEILLEURS RÉSULTATS DBSCAN")
    print("="*100)
    
    if not dbscan_results.empty:
        # Tri par silhouette_score décroissant
        top5_dbscan_sil = dbscan_results.nlargest(5, 'silhouette_score')
        print("\nTOP 5 - Silhouette Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_dbscan_sil.iterrows(), 1):
            print(f"{idx}. Score Silhouette: {row['silhouette_score']:.4f}")
            print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
            print(f"   Configuration DBSCAN: eps={row['eps']}, min_samples={row['min_samples']}")
            print(f"   Clusters détectés: {row['n_clusters_detectes']}, Points de bruit: {row['noise_points']}")
            print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
            print(f"   Échantillons: {row['n_samples']}")
            print()
        
        # Tri par davies_bouldin_score croissant
        top5_dbscan_db = dbscan_results.nsmallest(5, 'davies_bouldin_score')
        print("TOP 5 - Davies-Bouldin Score (meilleur = plus faible):")
        for idx, (i, row) in enumerate(top5_dbscan_db.iterrows(), 1):
            print(f"{idx}. Score Davies-Bouldin: {row['davies_bouldin_score']:.4f}")
            print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
            print(f"   Configuration DBSCAN: eps={row['eps']}, min_samples={row['min_samples']}")
            print(f"   Clusters détectés: {row['n_clusters_detectes']}, Points de bruit: {row['noise_points']}")
            print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
            print(f"   Échantillons: {row['n_samples']}")
            print()
        
        # Tri par calinski_harabasz_score décroissant
        top5_dbscan_ch = dbscan_results.nlargest(5, 'calinski_harabasz_score')
        print("TOP 5 - Calinski-Harabasz Score (meilleur = plus élevé):")
        for idx, (i, row) in enumerate(top5_dbscan_ch.iterrows(), 1):
            print(f"{idx}. Score Calinski-Harabasz: {row['calinski_harabasz_score']:.2f}")
            print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
            print(f"   Configuration DBSCAN: eps={row['eps']}, min_samples={row['min_samples']}")
            print(f"   Clusters détectés: {row['n_clusters_detectes']}, Points de bruit: {row['noise_points']}")
            print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
            print(f"   Échantillons: {row['n_samples']}")
            print()
    
    print("\n" + "="*100)
    print("TOP 5 MEILLEURS MODÈLES GLOBAUX")
    print("="*100)
    
    # Normalisation des scores pour comparaison inter-modèles
    # Création d'un score composite basé sur les 3 métriques
    df_resultats_copy = df_resultats.copy()
    
    # Normalisation min-max des scores entre 0 et 1 pour pouvoir les comparer
    df_resultats_copy['sil_norm'] = (df_resultats_copy['silhouette_score'] - df_resultats_copy['silhouette_score'].min()) / (df_resultats_copy['silhouette_score'].max() - df_resultats_copy['silhouette_score'].min())
    df_resultats_copy['db_norm'] = 1 - ((df_resultats_copy['davies_bouldin_score'] - df_resultats_copy['davies_bouldin_score'].min()) / (df_resultats_copy['davies_bouldin_score'].max() - df_resultats_copy['davies_bouldin_score'].min()))
    df_resultats_copy['ch_norm'] = (df_resultats_copy['calinski_harabasz_score'] - df_resultats_copy['calinski_harabasz_score'].min()) / (df_resultats_copy['calinski_harabasz_score'].max() - df_resultats_copy['calinski_harabasz_score'].min())
    
    # Score composite : moyenne pondérée des 3 métriques normalisées
    # Pondération: Silhouette 40%, Davies-Bouldin 30%, Calinski-Harabasz 30%
    df_resultats_copy['score_composite'] = (df_resultats_copy['sil_norm'] * 0.4 + 
                                           df_resultats_copy['db_norm'] * 0.3 + 
                                           df_resultats_copy['ch_norm'] * 0.3)
    
    top5_global = df_resultats_copy.nlargest(5, 'score_composite')
    print("\nTOP 5 - Score Composite (silhouette 40% + davies-bouldin 30% + calinski-harabasz 30%):")
    for idx, (i, row) in enumerate(top5_global.iterrows(), 1):
        config_clustering = ""
        if row['modele_clustering'] == 'KMeans':
            config_clustering = f"K-Means avec k_clusters={row['k_clusters']}"
        else:
            config_clustering = f"DBSCAN avec eps={row['eps']}, min_samples={row['min_samples']}"
        
        print(f"{idx}. Score Composite: {row['score_composite']:.4f}")
        print(f"   Modèle: {config_clustering}")
        print(f"   Configuration NN: k={row['nn_k']}, métrique={row['nn_metric']}, algorithme={row['nn_algorithm']}")
        print(f"   Prétraitement: {row['scaler']}, Cas: {row['cas']}")
        print(f"   Scores individuels: Sil={row['silhouette_score']:.4f}, DB={row['davies_bouldin_score']:.4f}, CH={row['calinski_harabasz_score']:.2f}")
        print()
    
    # ANALYSE GLOBALE DES PARAMÈTRES INFLUENTS
    print("\n" + "="*100)
    print("ANALYSE GLOBALE DES PARAMÈTRES INFLUENTS")
    print("="*100)
    
    # Analyse par cas de features
    print("\nINFLUENCE DU NOMBRE DE FEATURES:")
    for cas in df_resultats['cas'].unique():
        subset = df_resultats[df_resultats['cas'] == cas]
        avg_sil = subset['silhouette_score'].mean()
        std_sil = subset['silhouette_score'].std()
        count = len(subset)
        print(f"- {cas} ({subset.iloc[0]['features_count']} features): "
              f"Silhouette moyen = {avg_sil:.4f} ± {std_sil:.4f} (sur {count} tests)")
    
    # Analyse par scaler
    print("\nINFLUENCE DU TYPE DE PREPROCESSING:")
    for scaler in df_resultats['scaler'].unique():
        subset = df_resultats[df_resultats['scaler'] == scaler]
        avg_sil = subset['silhouette_score'].mean()
        std_sil = subset['silhouette_score'].std()
        count = len(subset)
        print(f"- {scaler}: Silhouette moyen = {avg_sil:.4f} ± {std_sil:.4f} (sur {count} tests)")
    
    # Analyse par métrique de distance NN
    print("\nINFLUENCE DE LA MÉTRIQUE DE DISTANCE NEARESTNEIGHBORS:")
    for metric in df_resultats['nn_metric'].unique():
        subset = df_resultats[df_resultats['nn_metric'] == metric]
        avg_sil = subset['silhouette_score'].mean()
        std_sil = subset['silhouette_score'].std()
        count = len(subset)
        print(f"- {metric}: Silhouette moyen = {avg_sil:.4f} ± {std_sil:.4f} (sur {count} tests)")
    
    # Analyse par algorithme NN
    print("\nINFLUENCE DE L'ALGORITHME NEARESTNEIGHBORS:")
    for algo in df_resultats['nn_algorithm'].unique():
        subset = df_resultats[df_resultats['nn_algorithm'] == algo]
        avg_sil = subset['silhouette_score'].mean()
        std_sil = subset['silhouette_score'].std()
        count = len(subset)
        print(f"- {algo}: Silhouette moyen = {avg_sil:.4f} ± {std_sil:.4f} (sur {count} tests)")
    
    # Analyse par modèle de clustering
    print("\nPERFORMANCE COMPARATIVE DES MODÈLES DE CLUSTERING:")
    for modele in df_resultats['modele_clustering'].unique():
        subset = df_resultats[df_resultats['modele_clustering'] == modele]
        avg_sil = subset['silhouette_score'].mean()
        avg_db = subset['davies_bouldin_score'].mean()
        avg_ch = subset['calinski_harabasz_score'].mean()
        count = len(subset)
        print(f"- {modele}: Sil={avg_sil:.4f}, DB={avg_db:.4f}, CH={avg_ch:.2f} (sur {count} tests)")
    
    # EXPLICATION DES MÉTRIQUES D'ÉVALUATION
    print("\n" + "="*100)
    print("EXPLICATION DES MÉTRIQUES D'ÉVALUATION")
    print("="*100)
    
    print("\n1. SILHOUETTE SCORE:")
    print("   UTILITÉ: Mesure la cohésion intra-cluster vs séparation inter-cluster")
    print("   INTERPRÉTATION: Valeurs entre -1 et 1, optimal proche de 1")
    print("   AVANTAGE: Indique si les points sont bien assignés à leur cluster")
    print("   LIMITATION: Pénalise les clusters de formes non-sphériques")
    print("   USAGE: Métrique principale pour évaluer la qualité globale du clustering")
    
    print("\n2. DAVIES-BOULDIN SCORE:")
    print("   UTILITÉ: Mesure la similarité moyenne entre chaque cluster et son plus similaire")
    print("   INTERPRÉTATION: Plus la valeur est faible, meilleur est le clustering")
    print("   AVANTAGE: Évalue la compacité intra-cluster et la séparation inter-cluster")
    print("   LIMITATION: Particulièrement sensible aux clusters de formes sphériques")
    print("   USAGE: Complément au silhouette score pour valider la qualité")
    
    print("\n3. CALINSKI-HARABASZ SCORE:")
    print("   UTILITÉ: Ratio de la variance inter-cluster sur la variance intra-cluster")
    print("   INTERPRÉTATION: Plus la valeur est élevée, meilleur est le clustering")
    print("   AVANTAGE: Favorise les clusters compacts et bien séparés")
    print("   LIMITATION: Peut être biaisé par le nombre de clusters")
    print("   USAGE: Aussi appelé Variance Ratio Criterion, utile pour comparer différents k")
    
    print("\n" + "="*100)
    print("RECOMMANDATIONS POUR LE DÉPLOIEMENT")
    print("="*100)
    
    best_model = top5_global.iloc[0]
    print(f"\nMEILLEUR MODÈLE RECOMMANDÉ POUR DÉPLOIEMENT:")
    print(f"- Modèle de clustering: {best_model['modele_clustering']}")
    print(f"- Cas de features: {best_model['cas']} ({best_model['features_count']} features)")
    print(f"- Features utilisées: {best_model['features_list']}")
    print(f"- Préprocessing: {best_model['scaler']}")
    print(f"- Configuration NearestNeighbors: k={best_model['nn_k']}, métrique={best_model['nn_metric']}, algorithme={best_model['nn_algorithm']}")
    
    if best_model['modele_clustering'] == 'KMeans':
        print(f"- Paramètres K-Means: nombre de clusters = {best_model['k_clusters']}")
        print(f"\nCODE DE DÉPLOIEMENT RECOMMANDÉ:")
        print(f"# Configuration optimale identifiée")
        print(f"from sklearn.preprocessing import {best_model['scaler']}")
        print(f"from sklearn.cluster import KMeans")
        print(f"from sklearn.neighbors import NearestNeighbors")
        print(f"from sklearn.pipeline import Pipeline")
        print(f"")
        print(f"# Pipeline de preprocessing et clustering")
        print(f"features = {best_model['features_list'].split(', ')}")
        print(f"pipeline = Pipeline([")
        print(f"    ('scaler', {best_model['scaler']}()),")
        print(f"    ('kmeans', KMeans(n_clusters={best_model['k_clusters']}, random_state=42))")
        print(f"])")
        print(f"")
        print(f"# NearestNeighbors pour recommandations")
        print(f"nn_model = NearestNeighbors(")
        print(f"    n_neighbors={best_model['nn_k']},")
        print(f"    metric='{best_model['nn_metric']}',")
        print(f"    algorithm='{best_model['nn_algorithm']}'")
        print(f")")
    else:
        print(f"- Paramètres DBSCAN: eps={best_model['eps']}, min_samples={best_model['min_samples']}")
        print(f"- Clusters détectés: {best_model.get('n_clusters_detectes', 'N/A')}")
        print(f"\nCODE DE DÉPLOIEMENT RECOMMANDÉ:")
        print(f"# Configuration optimale identifiée")
        print(f"from sklearn.preprocessing import {best_model['scaler']}")
        print(f"from sklearn.cluster import DBSCAN")
        print(f"from sklearn.neighbors import NearestNeighbors")
        print(f"from sklearn.pipeline import Pipeline")
        print(f"")
        print(f"# Pipeline de preprocessing et clustering")
        print(f"features = {best_model['features_list'].split(', ')}")
        print(f"pipeline = Pipeline([")
        print(f"    ('scaler', {best_model['scaler']}()),")
        print(f"    ('dbscan', DBSCAN(eps={best_model['eps']}, min_samples={best_model['min_samples']}))")
        print(f"])")
        print(f"")
        print(f"# NearestNeighbors pour recommandations")
        print(f"nn_model = NearestNeighbors(")
        print(f"    n_neighbors={best_model['nn_k']},")
        print(f"    metric='{best_model['nn_metric']}',")
        print(f"    algorithm='{best_model['nn_algorithm']}'")
        print(f")")
    
    print(f"")
    print(f"# Métriques de performance attendues:")
    print(f"# Silhouette Score: {best_model['silhouette_score']:.4f}")
    print(f"# Davies-Bouldin Score: {best_model['davies_bouldin_score']:.4f}")
    print(f"# Calinski-Harabasz Score: {best_model['calinski_harabasz_score']:.2f}")
    
    print("\n" + "="*100)
    print("ANALYSE PRÉDICTIVE ET CONSEILS DE DÉPLOIEMENT")
    print("="*100)
    
    print("\nPRÉDICTION DE PERFORMANCE:")
    print("- Les modèles avec un score composite > 0.7 sont considérés comme excellents")
    print("- Les modèles avec un score composite > 0.5 sont considérés comme bons")
    print("- Les modèles avec un score composite < 0.3 sont à éviter en production")
    print(f"- Le meilleur modèle obtient un score composite de {best_model['score_composite']:.4f}")
    
    print("\nPARAMÈTRES LES PLUS INFLUENTS (par ordre d'importance):")
    # Calcul de l'influence des paramètres sur le score composite
    correlations = []
    for param in ['cas', 'scaler', 'nn_metric', 'nn_algorithm']:
        param_influence = df_resultats_copy.groupby(param)['score_composite'].mean().std()
        correlations.append((param, param_influence))
    
    correlations.sort(key=lambda x: x[1], reverse=True)
    for i, (param, influence) in enumerate(correlations, 1):
        print(f"{i}. {param.upper()}: variance = {influence:.4f}")
    
    print("\nCONSEILS DE DÉPLOIEMENT EN PRODUCTION:")
    print("1. SURVEILLANCE CONTINUE:")
    print("   - Monitorer les métriques de performance sur de nouvelles données")
    print("   - Recalculer périodiquement les scores d'évaluation")
    print("   - Détecter la dérive des données (data drift)")
    
    print("\n2. OPTIMISATION MÉMOIRE:")
    print("   - Utiliser des échantillons représentatifs pour l'entraînement")
    print("   - Implémenter une mise en cache des résultats fréquents")
    print("   - Optimiser les calculs de distances pour NearestNeighbors")
    
    print("\n3. INTÉGRATION APPLICATIVE:")
    print("   - Sauvegarder le pipeline complet avec pickle ou joblib")
    print("   - Implémenter une API REST pour les prédictions")
    print("   - Prévoir un système de fallback en cas d'erreur")
    
    print("\n4. ÉVOLUTIVITÉ:")
    print("   - Prévoir un réentraînement périodique automatique")
    print("   - Implémenter un système de validation A/B")
    print("   - Garder les anciennes versions du modèle pour rollback")
    
    # Sauvegarde des résultats
    try:
        df_resultats.to_csv('resultats_ml_clustering_complets.csv', index=False)
        top5_global.to_csv('top5_meilleurs_modeles_ml.csv', index=False)
        print(f"\nRésultats sauvegardés dans:")
        print(f"- resultats_ml_clustering_complets.csv")
        print(f"- top5_meilleurs_modeles_ml.csv")
    except Exception as e:
        print(f"Erreur lors de la sauvegarde: {e}")
    
    return {
        'tous_resultats': df_resultats,
        'top5_global': top5_global,
        'meilleur_modele': best_model,
        'top5_kmeans': {'silhouette': top5_kmeans_sil, 'davies_bouldin': top5_kmeans_db, 'calinski_harabasz': top5_kmeans_ch} if not kmeans_results.empty else {},
        'top5_dbscan': {'silhouette': top5_dbscan_sil, 'davies_bouldin': top5_dbscan_db, 'calinski_harabasz': top5_dbscan_ch} if not dbscan_results.empty else {}
    }

def generer_rapport_texte(analyses, nom_fichier='rapport_analyse_clustering_musical.txt'):
    """
    Génère un rapport texte complet de l'analyse (alternative au PDF)
    
    Arguments:
    analyses : Dictionnaire contenant toutes les analyses
    nom_fichier : Nom du fichier texte à générer
    """
    
    if not analyses:
        print("Aucune analyse à inclure dans le rapport")
        return
    
    try:
        with open(nom_fichier, 'w', encoding='utf-8') as f:
            # En-tête du rapport
            f.write("="*80 + "\n")
            f.write("RAPPORT D'ANALYSE - MODELES DE CLUSTERING MUSICAL\n")
            f.write("="*80 + "\n\n")
            
            # Introduction
            f.write("1. RESUME EXECUTIF\n")
            f.write("-"*50 + "\n")
            f.write("Cette analyse présente une évaluation exhaustive de modèles de clustering non supervisé "
                   "appliqués à des données musicales. L'objectif est d'identifier les meilleures configurations "
                   "pour segmenter automatiquement des morceaux de musique selon leurs caractéristiques audio.\n\n")
            
            # Méthodologie
            f.write("2. METHODOLOGIE\n")
            f.write("-"*50 + "\n")
            f.write("L'analyse a testé systématiquement toutes les combinaisons possibles de paramètres :\n")
            f.write("• 3 cas de features audio (5, 6 et 7 caractéristiques)\n")
            f.write("• 3 méthodes de préprocessing (StandardScaler, MinMaxScaler, RobustScaler)\n")
            f.write("• Paramètres NearestNeighbors variables (k=1 à 25, métriques euclidean/manhattan)\n")
            f.write("• 2 algorithmes de clustering (K-Means et DBSCAN)\n")
            f.write("• 3 métriques d'évaluation (Silhouette, Davies-Bouldin, Calinski-Harabasz)\n\n")
            
            # Métriques d'évaluation
            f.write("3. METRIQUES D'EVALUATION\n")
            f.write("-"*50 + "\n")
            f.write("3.1 Silhouette Score\n")
            f.write("Mesure la cohésion intra-cluster versus la séparation inter-cluster. "
                   "Valeurs entre -1 et 1, avec 1 indiquant un clustering parfait. "
                   "Métrique principale pour évaluer la qualité globale.\n\n")
            
            f.write("3.2 Davies-Bouldin Score\n")
            f.write("Évalue la similarité moyenne entre chaque cluster et son plus similaire. "
                   "Plus la valeur est faible, meilleur est le clustering. "
                   "Complément important au Silhouette Score.\n\n")
            
            f.write("3.3 Calinski-Harabasz Score\n")
            f.write("Ratio de la variance inter-cluster sur la variance intra-cluster. "
                   "Plus la valeur est élevée, plus les clusters sont distincts. "
                   "Particulièrement utile pour comparer différents nombres de clusters.\n\n")
            
            # Résultats Top 5
            if 'top5_global' in analyses and not analyses['top5_global'].empty:
                f.write("4. RESULTATS - TOP 5 MODELES GLOBAUX\n")
                f.write("-"*50 + "\n")
                
                top5 = analyses['top5_global']
                for idx, (i, row) in enumerate(top5.iterrows(), 1):
                    if row['modele_clustering'] == 'KMeans':
                        config = f"K-Means (k={row['k_clusters']})"
                    else:
                        config = f"DBSCAN (eps={row['eps']}, min_samples={row['min_samples']})"
                    
                    f.write(f"{idx}. Score: {row['score_composite']:.4f} | {config} | "
                           f"NN: k={row['nn_k']}, {row['nn_metric']} | "
                           f"Scaler: {row['scaler']} | Cas: {row['cas']}\n")
                f.write("\n")
            
            # Recommandations
            f.write("5. RECOMMANDATIONS DE DEPLOIEMENT\n")
            f.write("-"*50 + "\n")
            
            if 'meilleur_modele' in analyses:
                best = analyses['meilleur_modele']
                
                f.write("5.1 Configuration optimale recommandée\n")
                f.write(f"• Modèle: {best['modele_clustering']}\n")
                f.write(f"• Cas de features: {best['cas']} ({best['features_count']} caractéristiques)\n")
                f.write(f"• Preprocessing: {best['scaler']}\n")
                f.write(f"• Configuration NearestNeighbors: k={best['nn_k']}, métrique={best['nn_metric']}\n")
                
                if best['modele_clustering'] == 'KMeans':
                    f.write(f"• Paramètres K-Means: {best['k_clusters']} clusters\n")
                else:
                    f.write(f"• Paramètres DBSCAN: eps={best['eps']}, min_samples={best['min_samples']}\n")
                
                f.write(f"• Score composite obtenu: {best['score_composite']:.4f}\n\n")
            
            f.write("5.2 Conseils d'implémentation\n")
            f.write("• Sauvegarder le pipeline complet (preprocessing + clustering) avec joblib\n")
            f.write("• Implémenter une surveillance continue des métriques de performance\n")
            f.write("• Prévoir un système de réentraînement périodique automatique\n")
            f.write("• Utiliser des échantillons représentatifs pour optimiser les performances\n")
            f.write("• Mettre en place un système de validation A/B pour les évolutions\n\n")
            
            # Conclusion
            f.write("6. CONCLUSION\n")
            f.write("-"*50 + "\n")
            f.write("Cette analyse exhaustive a permis d'identifier les configurations optimales pour le "
                   "clustering de données musicales. Les résultats démontrent l'importance du choix des "
                   "paramètres de preprocessing et de la configuration des algorithmes. Le modèle recommandé "
                   "offre un excellent équilibre entre performance et stabilité pour un déploiement en production.\n")
        
        print(f"\nRapport texte généré avec succès: {nom_fichier}")
        
    except Exception as e:
        print(f"Erreur lors de la génération du rapport texte: {e}")

def executer_analyse_complete():
    """
    Fonction principale pour exécuter l'analyse complète des modèles ML
    
    Cette fonction orchestre l'ensemble du processus:
    1. Charge les données depuis le fichier CSV
    2. Teste toutes les configurations possibles
    3. Analyse les résultats et génère les classements
    4. Produit le rapport texte final (alternative au PDF)
    
    Retourne:
    dict : Résultats complets de l'analyse
    """
    
    print("DÉBUT DE L'ANALYSE EXHAUSTIVE DES MODÈLES ML DE CLUSTERING MUSICAL")
    print("="*100)
    print(f"Système d'exploitation détecté: {platform.system()}")
    
    try:
        # Chargement des données
        df_spotify = charger_donnees()
        
        # Exécution de tous les tests de configurations
        resultats = test_configurations_ml_complet(df_spotify)
        
        if not resultats:
            print("ERREUR: Aucun résultat généré lors des tests")
            return None
        
        print(f"\nNOMBRE TOTAL DE CONFIGURATIONS TESTÉES: {len(resultats)}")
        
        # Analyse complète des résultats et génération des classements
        analyses = analyser_et_classer_resultats(resultats)
        
        # Génération du rapport texte (alternative au PDF)
        generer_rapport_texte(analyses)
        
        print("\nANALYSE COMPLÈTE TERMINÉE AVEC SUCCÈS!")
        print("Tous les fichiers de résultats ont été générés:")
        print("- Fichiers CSV avec les résultats détaillés")
        print("- Rapport texte professionnel")
        print("\nNOTE: Pour générer un PDF, installez fpdf2 avec: pip install fpdf2")
        
        return analyses
        
    except Exception as e:
        print(f"ERREUR LORS DE L'EXÉCUTION DE L'ANALYSE: {e}")
        import traceback
        traceback.print_exc()
        return None

# EXÉCUTION AUTOMATIQUE DE L'ANALYSE COMPLÈTE
if __name__ == "__main__":
    resultats_finaux = executer_analyse_complete()
```

class MoodifyRecommender:
    """
    Système de recommandation basé sur votre modèle de clustering ML entraîné
    """
    
    def __init__(self, model_path, data_path):
        """
        Initialise le système de recommandation
        
        Arguments:
        model_path : Chemin vers votre fichier .pkl du modèle entraîné
        data_path : Chemin vers vos données Spotify complètes
        """
        self.model_info = self.charger_modele(model_path)
        self.df_songs = pd.read_csv(data_path)
        self.features = self.model_info['features'] if self.model_info else []
        self.songs_processed = None
        self.clusters_assigned = None
        
        if self.model_info:
            self.preparer_donnees()
    
    def charger_modele(self, model_path):
        """
        Charge votre modèle ML entraîné
        
        Arguments:
        model_path : Chemin vers le fichier .pkl
        
        Retourne:
        dict : Informations du modèle ou None si erreur
        """
        try:
            with open(model_path, 'rb') as f:
                model_info = pickle.load(f)
            return model_info
        except Exception as e:
            st.error(f"Erreur lors du chargement du modèle: {e}")
            return None
    
    def preparer_donnees(self):
        """
        Prépare les données avec le même preprocessing que l'entraînement
        """
        try:
            # Sélection des features utilisées par le modèle
            self.songs_processed = self.df_songs[self.features].copy()
            
            # Nettoyage des données
            self.songs_processed = self.songs_processed.dropna()
            
            # Application du preprocessing avec le pipeline du modèle
            self.clusters_assigned = self.model_info['pipeline'].predict(self.songs_processed)
            
            # Ajout des clusters au DataFrame
            self.df_songs_with_clusters = self.df_songs.iloc[self.songs_processed.index].copy()
            self.df_songs_with_clusters['cluster'] = self.clusters_assigned
            
            # Préparation des données scalées pour le calcul de similarité
            self.songs_scaled = self.model_info['pipeline'].named_steps['scaler'].transform(self.songs_processed)
            
        except Exception as e:
            st.error(f"Erreur lors de la préparation des données: {e}")
    
    def rechercher_chanson(self, query, search_by='track_name'):
        """
        Recherche une chanson dans le dataset
        
        Arguments:
        query : Terme de recherche
        search_by : Colonne pour la recherche ('track_name', 'artist_name', etc.)
        
        Retourne:
        DataFrame : Chansons correspondantes
        """
        if search_by not in self.df_songs_with_clusters.columns:
            return pd.DataFrame()
        
        # Recherche insensible à la casse
        mask = self.df_songs_with_clusters[search_by].str.contains(query, case=False, na=False)
        results = self.df_songs_with_clusters[mask]
        
        return results.head(10)  # Limiter à 10 résultats
    
    def recommander_songs_similaires(self, song_index, n_recommendations=5, method='cluster_cosine'):
        """
        Recommande des chansons similaires basées sur votre modèle ML
        
        Arguments:
        song_index : Index de la chanson de référence
        n_recommendations : Nombre de recommandations
        method : Méthode de recommandation ('cluster_cosine', 'cluster_euclidean', 'global_cosine')
        
        Retourne:
        DataFrame : Chansons recommandées
        """
        if song_index not in self.df_songs_with_clusters.index:
            return pd.DataFrame()
        
        song_cluster = self.df_songs_with_clusters.loc[song_index, 'cluster']
        song_features = self.songs_scaled[self.songs_processed.index.get_loc(song_index)].reshape(1, -1)
        
        if method == 'cluster_cosine':
            # Recommandations basées sur le cluster + similarité cosine
            cluster_songs = self.df_songs_with_clusters[
                (self.df_songs_with_clusters['cluster'] == song_cluster) & 
                (self.df_songs_with_clusters.index != song_index)
            ]
            
            if len(cluster_songs) == 0:
                return pd.DataFrame()
            
            # Calcul de similarité cosine dans le cluster
            cluster_indices = [self.songs_processed.index.get_loc(idx) for idx in cluster_songs.index]
            cluster_features = self.songs_scaled[cluster_indices]
            
            similarities = cosine_similarity(song_features, cluster_features)[0]
            
        elif method == 'cluster_euclidean':
            # Recommandations basées sur le cluster + distance euclidienne
            cluster_songs = self.df_songs_with_clusters[
                (self.df_songs_with_clusters['cluster'] == song_cluster) & 
                (self.df_songs_with_clusters.index != song_index)
            ]
            
            if len(cluster_songs) == 0:
                return pd.DataFrame()
            
            cluster_indices = [self.songs_processed.index.get_loc(idx) for idx in cluster_songs.index]
            cluster_features = self.songs_scaled[cluster_indices]
            
            distances = euclidean_distances(song_features, cluster_features)[0]
            similarities = 1 / (1 + distances)  # Conversion en similarité
            
        else:  # global_cosine
            # Recommandations sur tout le dataset
            cluster_songs = self.df_songs_with_clusters[self.df_songs_with_clusters.index != song_index]
            
            all_indices = [self.songs_processed.index.get_loc(idx) for idx in cluster_songs.index]
            all_features = self.songs_scaled[all_indices]
            
            similarities = cosine_similarity(song_features, all_features)[0]
        
        # Tri par similarité décroissante
        similar_indices = np.argsort(similarities)[::-1][:n_recommendations]
        recommended_songs = cluster_songs.iloc[similar_indices].copy()
        recommended_songs['similarity_score'] = similarities[similar_indices]
        
        return recommended_songs
    
    def get_song_profile(self, song_index):
        """
        Obtient le profil détaillé d'une chanson
        
        Arguments:
        song_index : Index de la chanson
        
        Retourne:
        dict : Profil de la chanson
        """
        if song_index not in self.df_songs_with_clusters.index:
            return {}
        
        song_data = self.df_songs_with_clusters.loc[song_index]
        song_cluster = song_data['cluster']
        
        # Profil du cluster
        cluster_profile = self.model_info.get('cluster_profiles', {}).get(song_cluster, {})
        
        # Features de la chanson
        song_features = {feature: song_data.get(feature, 0) for feature in self.features}
        
        return {
            'song_info': song_data.to_dict(),
            'cluster': song_cluster,
            'cluster_profile': cluster_profile,
            'features': song_features
        }
